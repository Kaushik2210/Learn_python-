## ① Basics

```python
# Types & conversion
print(type(5), type(5.0), type(5j), type(True), type("s"), type(None), type([]), type(()), type({}), type({1}))
print(int("42"), float("3.5"), str(9), bool(""), list("ab"), tuple([1]), set([1, 1]))
# Operators: arithmetic   + - * / // % **        comparison  == != < > <= >=
#            logical      and or not             bitwise     & | ^ ~ << >>
#            identity     is  is not             membership  in  not in
print(7 / 2, 7 // 2, 7 % 2, 2 ** 5, -7 // 2, -7 % 2, divmod(7, 2))
print(5 & 3, 5 | 3, 5 ^ 3, ~5, 5 << 1, 5 >> 1)
# Precedence: ()  **  unary  * / // %  + -  shifts  &  ^  |  comparisons  not  and  or
print(2 + 3 * 4 ** 2 / 8, not True or True, 2 ** 3 ** 2)
# Falsy: 0 0.0 0j "" [] () {} set() None False
print([bool(x) for x in (0, 0.0, "", [], (), {}, None, "0", [0])])
# Input / output
name, age = "Asha", 19
print(f"{name} is {age}", "a", "b", sep="-", end="!\n")
print("%s is %d years, %.2f" % (name, age, 3.14159), "{} {}".format(name, age))
```

## ② Strings

## ③ Lists

```python
L = [3, 1, 2]
L.append(4); L.extend([5, 6]); L.insert(0, 0)
print(L, L.pop(), L.pop(0), L)
L.remove(3); L.sort(); print(L, L.index(2), L.count(1)); L.reverse(); print(L)
print(sorted(L), sorted(L, reverse=True), len(L), sum(L), min(L), max(L))
print(L[:2], L[::-1], L + [9], L * 2, 2 in L)
del L[0]; print(L); L.clear(); print(L)
# append(x) adds ONE item; extend(it) adds each item of it; sort() returns None; copy: L[:], L.copy(), list(L)
a = [1, 2]; b = a; c = a[:]; b.append(3); print(a, c)
print([x * x for x in range(5) if x % 2 == 0])
```

## ⑤ Control flow

```python
n = 7
if n > 10: r = "big"
elif n > 5: r = "medium"
else: r = "small"
print(r, "even" if n % 2 == 0 else "odd")
for i in range(1, 10, 3): print(i, end=" ")            # range(start, stop, step): stop excluded
print()
i = 0
while i < 3:
    i += 1
    if i == 2: continue        # skip rest of this iteration
    print("i =", i)
else:
    print("while-else runs when the loop did not break")
for x in range(3):
    if x == 5: break
else:
    print("for-else: no break happened")
for idx, ch in enumerate("ab", 1): print(idx, ch)
for p, q in zip([1, 2], "xy"): print(p, q)
```

## ⑥ Functions

```python
def f(a, b=2, *args, key="k", **kwargs):
    """Docstring."""
    return a, b, args, key, kwargs
print(f(1), f(1, 9, 8, 7, key="z", extra=0))
sq = lambda x: x * x
print(sq(4), list(map(sq, [1, 2])), list(filter(lambda v: v > 1, [1, 2, 3])))
from functools import reduce
print(reduce(lambda a, b: a + b, [1, 2, 3]))
# LEGB scope: Local -> Enclosing -> Global -> Built-in   (global / nonlocal keywords)
# Mutable default trap: def g(x, lst=[])  -> use None
# Return multiple values = tuple. No return = None.
def fact(n): return 1 if n <= 1 else n * fact(n - 1)
print(fact(5))
# Decorator
def deco(fn):
    def wrap(*a, **k):
        print("before"); r = fn(*a, **k); print("after"); return r
    return wrap
@deco
def hi(): return "hi"
print(hi())
# Generator
def gen(n):
    for i in range(n): yield i
print(list(gen(3)), sum(x for x in range(4)))
```

## ⑦ OOP

```python
class Animal:
    count = 0                                  # class attribute
    def __init__(self, name):                  # constructor
        self.name = name                       # instance attribute
        self.__secret = 1                      # private (name-mangled)
        Animal.count += 1
    def speak(self): return f"{self.name} makes a sound"
    def __str__(self): return f"Animal({self.name})"
    def __repr__(self): return f"Animal({self.name!r})"
    def __eq__(self, o): return self.name == o.name
    def __len__(self): return len(self.name)
    def __add__(self, o): return Animal(self.name + o.name)
    @property
    def upper(self): return self.name.upper()
    @classmethod
    def make(cls, n): return cls(n)
    @staticmethod
    def kind(): return "animal"
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name); self.breed = breed
    def speak(self): return f"{self.name} barks"      # overriding
d = Dog("Rex", "lab")
print(d.speak(), str(d), repr(d), len(d), d.upper, Animal.make("x"), Animal.kind(), Animal.count)
print(isinstance(d, Animal), issubclass(Dog, Animal), Dog.__mro__[1].__name__)
# Pillars: Encapsulation, Abstraction (abc), Inheritance, Polymorphism (overriding, duck typing)
# Types of inheritance: single, multiple, multilevel, hierarchical, hybrid.  MRO = C3 linearization.
```

## ⑧ Exceptions

```python
def demo(x):
    try:
        r = 10 / x
    except ZeroDivisionError as e:
        return f"error: {e}"
    except (TypeError, ValueError):
        return "type/value error"
    else:
        return f"ok {r}"            # runs only if no exception
    finally:
        print("finally always runs")
print(demo(2), demo(0), demo("a"))
# raise ValueError("msg");  assert cond, "msg";  class MyErr(Exception): pass
# Hierarchy: BaseException > Exception > (ArithmeticError > ZeroDivisionError), (LookupError > IndexError, KeyError),
#            ValueError, TypeError, NameError, AttributeError, OSError > FileNotFoundError, ImportError, RuntimeError ...
```

```python
# Quick self-test: can you write these 5 in under 5 minutes, without looking?
def is_prime(n): return n > 1 and all(n % d for d in range(2, int(n ** 0.5) + 1))
def fib(n):
    a, b = 0, 1
    for _ in range(n): yield a; a, b = b, a + b
def is_pal(s): return (s := str(s)) == s[::-1]
def gcd(a, b): return a if not b else gcd(b, a % b)
def bubble(a):
    a = a[:]
    for i in range(len(a)):
        for j in range(len(a) - 1 - i):
            if a[j] > a[j + 1]: a[j], a[j + 1] = a[j + 1], a[j]
    return a
print([n for n in range(20) if is_prime(n)], list(fib(8)), is_pal(12321), gcd(84, 36), bubble([4, 2, 9, 1]))
```
