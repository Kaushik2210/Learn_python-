# 🐞 Exam Prep 05 — Find the Bug (16 debugging drills)

Debugging questions ("the following code has errors — correct them") are common in exams, and debugging is the skill that makes you fast in real life.

### Routine
1. `drill(n)` prints a buggy function and what it *should* do.
2. **Copy it into a new cell**, find the bug, fix it. Run `test(n, your_function)` to check.
3. Stuck? `hint(n)`. Give up? `solution(n)` shows the fix **and an explanation**.

Every drill was verified automatically: the buggy version *fails* its test and the fixed version *passes*.

```python
import textwrap

DRILLS = []

def add(title, spec, buggy, fixed, name, check, hint, why):
    """check(obj) -> bool receives the function/class defined by the code."""
    DRILLS.append(dict(title=title, spec=spec, buggy=textwrap.dedent(buggy).strip(),
                       fixed=textwrap.dedent(fixed).strip(), name=name, check=check, hint=hint, why=why))

add("Average of a list", "average(nums) returns the mean, or 0 for an empty list.",
"""
def average(nums):
    total = 0
    for n in nums:
        total += n
    return total / len(nums)
""", """
def average(nums):
    if not nums:
        return 0
    return sum(nums) / len(nums)
""", "average",
lambda f: f([1, 2, 3]) == 2 and f([]) == 0 and f([5]) == 5,
"What happens when the list is empty?",
"Dividing by `len([]) == 0` raises `ZeroDivisionError`. Always handle the empty case.")

add("Largest number", "largest(nums) returns the biggest number, including when all numbers are negative.",
"""
def largest(nums):
    big = 0
    for n in nums:
        if n > big:
            big = n
    return big
""", """
def largest(nums):
    big = nums[0]
    for n in nums:
        if n > big:
            big = n
    return big
""", "largest",
lambda f: f([1, 5, 3]) == 5 and f([-4, -2, -9]) == -2,
"What is the answer for [-4, -2, -9]?",
"Starting from `0` is wrong when every number is negative. Initialise with the first element (or `float('-inf')`).")

add("Count vowels", "count_vowels(s) counts vowels case-insensitively.",
"""
def count_vowels(s):
    count = 0
    for ch in s:
        if ch in "aeiou":
            count += 1
    return count
""", """
def count_vowels(s):
    count = 0
    for ch in s.lower():
        if ch in "aeiou":
            count += 1
    return count
""", "count_vowels",
lambda f: f("Hello") == 2 and f("AEIOU") == 5 and f("xyz") == 0,
"Try it with capital letters.",
"Uppercase vowels are not in `\"aeiou\"`; normalise with `.lower()`.")

add("Remove evens", "remove_evens(lst) returns the list without its even numbers (and must not skip any).",
"""
def remove_evens(lst):
    for x in lst:
        if x % 2 == 0:
            lst.remove(x)
    return lst
""", """
def remove_evens(lst):
    return [x for x in lst if x % 2 != 0]
""", "remove_evens",
lambda f: f([1, 2, 3, 4, 5, 6]) == [1, 3, 5] and f([2, 4, 6, 8]) == [] and f([1, 3]) == [1, 3],
"What does the loop do when you remove an element while iterating?",
"Removing items while iterating shifts the remaining ones so some are skipped. Build a new list (comprehension) or iterate over a copy.")

add("Sum of digits", "digit_sum(n) sums the digits of a positive integer.",
"""
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n / 10
    return total
""", """
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
""", "digit_sum",
lambda f: f(123) == 6 and f(9) == 9 and f(1000) == 1,
"Does `n` ever change?",
"`n / 10` computes a value but never assigns it, so the loop never ends. Use `n //= 10` (integer division and assignment).")

add("Factorial", "factorial(n) for n >= 0 (0! = 1).",
"""
def factorial(n):
    if n == 0:
        return 0
    return n * factorial(n - 1)
""", """
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
""", "factorial",
lambda f: f(0) == 1 and f(1) == 1 and f(5) == 120,
"What is the base-case value?",
"The base case must return `1` (the multiplicative identity), otherwise every product becomes 0.")

add("Palindrome", "is_palindrome(s) is True if s reads the same both ways.",
"""
def is_palindrome(s):
    return s == s.reverse()
""", """
def is_palindrome(s):
    return s == s[::-1]
""", "is_palindrome",
lambda f: f("madam") is True and f("hello") is False and f("") is True,
"Do strings have a reverse() method?",
"Strings have no `.reverse()` (only lists do, and it returns `None`). Reverse a string with slicing `s[::-1]`.")

add("Default list", "add_item(item, box=None) appends to the given list, or to a NEW list each call when none is given.",
"""
def add_item(item, box=[]):
    box.append(item)
    return box
""", """
def add_item(item, box=None):
    if box is None:
        box = []
    box.append(item)
    return box
""", "add_item",
lambda f: (f("a"), f("b"))[1] == ["b"] and f("c", ["x"]) == ["x", "c"],
"Call it twice with one argument each.",
"Default arguments are evaluated once, so the same list is reused across calls. Use `None` and create a new list inside.")

add("Merge two dicts", "merge(a, b) returns a NEW dict; b's values win; the inputs must not change.",
"""
def merge(a, b):
    a.update(b)
    return a
""", """
def merge(a, b):
    result = dict(a)
    result.update(b)
    return result
""", "merge",
lambda f: (lambda a, b: f(a, b) == {"x": 1, "y": 2} and a == {"x": 1})({"x": 1}, {"y": 2}),
"After calling merge, print the original dict.",
"`update` mutates in place. Copy first (or use `{**a, **b}` / `a | b`).")

add("Find index", "find(lst, target) returns the index of target or -1.",
"""
def find(lst, target):
    for i in range(len(lst)):
        if lst[i] == target:
            return i
        else:
            return -1
""", """
def find(lst, target):
    for i in range(len(lst)):
        if lst[i] == target:
            return i
    return -1
""", "find",
lambda f: f([5, 6, 7], 7) == 2 and f([5, 6, 7], 5) == 0 and f([], 1) == -1,
"When does the else branch run?",
"The `else: return -1` is inside the loop, so it returns at the first mismatch. Move `return -1` after the loop.")

add("Word count", "word_count(text) returns a dict of word frequencies (lowercase).",
"""
def word_count(text):
    counts = {}
    for w in text.lower().split():
        counts[w] += 1
    return counts
""", """
def word_count(text):
    counts = {}
    for w in text.lower().split():
        counts[w] = counts.get(w, 0) + 1
    return counts
""", "word_count",
lambda f: f("a b A") == {"a": 2, "b": 1} and f("") == {},
"What is counts['a'] the first time?",
"A missing key raises `KeyError`. Use `get(w, 0)`, `setdefault`, `defaultdict(int)` or `Counter`.")

add("Swap function", "swap(a, b) returns the values swapped as a tuple.",
"""
def swap(a, b):
    a = b
    b = a
    return a, b
""", """
def swap(a, b):
    a, b = b, a
    return a, b
""", "swap",
lambda f: f(1, 2) == (2, 1),
"After `a = b`, what is a?",
"The first assignment destroys the old `a`. Use tuple unpacking `a, b = b, a` or a temporary variable.")

add("Shared class attribute", "Each Cart must keep its own items.",
"""
class Cart:
    items = []
    def add(self, x):
        self.items.append(x)
        return len(self.items)
""", """
class Cart:
    def __init__(self):
        self.items = []
    def add(self, x):
        self.items.append(x)
        return len(self.items)
""", "Cart",
lambda C: (lambda a, b: (a.add("pen"), b.add("book"))[1] == 1)(C(), C()),
"Where should per-object data be created?",
"A mutable class attribute is shared by every instance. Create per-instance data in `__init__`.")

add("Check even", "is_even(n) returns True for even numbers.",
"""
def is_even(n):
    if n % 2 = 0:
        return True
    return False
""", """
def is_even(n):
    return n % 2 == 0
""", "is_even",
lambda f: f(4) is True and f(7) is False,
"Look at the comparison operator.",
"`=` assigns, `==` compares (a `SyntaxError`). Also, `return n % 2 == 0` is simpler than if/else returning booleans.")

add("Sum of text numbers", "total_price(prices) where prices are strings like '10', '5.5'.",
"""
def total_price(prices):
    total = 0
    for p in prices:
        total += p
    return total
""", """
def total_price(prices):
    total = 0
    for p in prices:
        total += float(p)
    return total
""", "total_price",
lambda f: f(["10", "5.5"]) == 15.5,
"What type does input() or a CSV give you?",
"Text must be converted with `float()`/`int()` before arithmetic; `0 + '10'` raises `TypeError`.")

add("Closure in a loop", "make_multipliers() returns [f0, f1, f2] where fi(x) = x * i.",
"""
def make_multipliers():
    return [lambda x: x * i for i in range(3)]
""", """
def make_multipliers():
    return [lambda x, i=i: x * i for i in range(3)]
""", "make_multipliers",
lambda f: [g(10) for g in f()] == [0, 10, 20],
"When is `i` looked up?",
"Closures capture the *variable*, not its value, so all lambdas see the final `i` (2). Bind the current value with a default argument `i=i`.")

def _passes(d, code):
    """Execute `code` and run the drill's check. Any exception counts as a failure."""
    ns = {}
    try:
        exec(code, ns)
        return bool(d["check"](ns[d["name"]]))
    except Exception:
        return False

def _verify():
    bad = []
    for d in DRILLS:
        if d["title"] == "Sum of digits":            # the buggy version loops forever, so don't run it
            if not _passes(d, d["fixed"]):
                bad.append(d["title"])
            continue
        if _passes(d, d["buggy"]) or not _passes(d, d["fixed"]):
            bad.append(d["title"])
    return bad

problems = _verify()
print(f"{len(DRILLS)} drills loaded; self-verification problems: {problems or 'none'}")
assert not problems

def drill(n):
    d = DRILLS[n - 1]
    print(f"🐞 Drill {n}: {d['title']}\nGoal: {d['spec']}\n")
    print(d["buggy"])

def hint(n):
    print("💡", DRILLS[n - 1]["hint"])

def solution(n):
    d = DRILLS[n - 1]
    print(f"✅ Fixed version — Drill {n}: {d['title']}\n")
    print(d["fixed"])
    print("\n📖 Why:", d["why"])

def test(n, obj):
    """Check YOUR fixed function/class for drill n."""
    d = DRILLS[n - 1]
    try:
        ok = bool(d["check"](obj))
    except Exception as e:
        ok = False
        print("(your code raised", type(e).__name__, ")")
    print("✅ PASS — nice fix!" if ok else f"❌ Not yet. Try hint({n}).")
```

```python
# Drills 1-8
for n in range(1, 9):
    drill(n)
    print("-" * 60)
```

```python
# Drills 9-16
for n in range(9, len(DRILLS) + 1):
    drill(n)
    print("-" * 60)
```

## 📚 The 12 bugs that cause 90 % of beginner problems
| # | Bug | Symptom | Fix |
|---|---|---|---|
| 1 | `=` vs `==` | `SyntaxError` / wrong branch | `==` to compare |
| 2 | Missing colon or wrong indent | `SyntaxError` / `IndentationError` | `:` after `if/for/def`, 4 spaces |
| 3 | `input()` is a string | `TypeError` or `'55'` instead of 10 | `int(input())` |
| 4 | Off-by-one in `range`/slices | missing last item | stop is exclusive |
| 5 | Modifying a list while looping | skipped items | iterate over a copy / comprehension |
| 6 | Mutable default argument | data leaks between calls | `None` sentinel |
| 7 | Forgetting `return` (prints instead) | result is `None` | `return value` |
| 8 | `list.sort()` result assigned | `None` | use `sorted()` or don't assign |
| 9 | Using a variable before assignment / typo | `NameError` | check spelling & scope |
| 10 | Integer vs true division | `2` vs `2.5` | `//` vs `/` |
| 11 | Shared references (`b = a`, `[[0]*3]*3`) | mysterious changes | `copy()`, comprehension |
| 12 | Catching too much (`except:`) | bugs hidden | catch specific exceptions |

### Debugging drill for real life
1. Reproduce → 2. Read the last line of the traceback → 3. `print(type(x), repr(x))` → 4. Bisect (comment out half) → 5. Explain the code to a rubber duck → 6. Search the exact error text.
