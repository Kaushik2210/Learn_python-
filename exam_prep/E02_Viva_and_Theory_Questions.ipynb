## 1. Fundamentals
**1. What is Python?**
> A high-level, interpreted, dynamically typed, general-purpose, multi-paradigm language created by Guido van Rossum (1991).

**2. Features of Python?**
> Simple/readable syntax, free & open source, portable, interpreted, dynamically typed, object-oriented + functional + procedural, huge standard library, extensible, automatic memory management.

**3. Is Python compiled or interpreted?**
> Both: source is compiled to bytecode (`.pyc`) which the Python Virtual Machine interprets.

**4. What is bytecode / PVM?**
> Bytecode is the platform-independent low-level instruction set; the PVM executes it.

**5. Difference between a script and a module?**
> Same thing physically (a `.py` file). A script is run directly; a module is imported. `__name__ == "__main__"` tells which.

**6. What is PEP 8?**
> Python's style guide (4-space indentation, snake_case, max ~79 chars per line, etc.).

**7. What is indentation and why is it important?**
> Leading whitespace defines blocks. Wrong indentation = `IndentationError`/logic errors.

**8. What are the different modes to run Python?**
> Interactive (REPL), script mode, and notebooks/IDEs.

**9. What is the difference between `python` and `ipython`/Jupyter?**
> IPython is an enhanced interactive shell (tab completion, magics); Jupyter provides browser/VS Code notebooks mixing code, output and text.

**10. What is CPython? Other implementations?**
> CPython is the reference implementation in C. Others: PyPy (JIT), Jython (JVM), IronPython (.NET), MicroPython.

## 2. Data types & operators
**11. Name the built-in data types.**
> Numeric (`int`, `float`, `complex`), `bool`, `str`, sequences (`list`, `tuple`, `range`), `set`/`frozenset`, `dict`, `bytes`/`bytearray`, `NoneType`.

**12. What is dynamic typing?**
> Types are attached to values, not variable names; a name can be rebound to any type and types are checked at run time.

**13. Mutable vs immutable types?**
> Mutable (can change in place): list, dict, set, bytearray. Immutable: int, float, complex, bool, str, tuple, frozenset, bytes, None.

**14. Difference between `/`, `//` and `%`?**
> True division (always float), floor division (rounds toward −∞), remainder (sign follows the divisor).

**15. What is `-7 // 2` and `-7 % 2`?**
> `-4` and `1` (because `-7 = (-4)*2 + 1`).

**16. Difference between `is` and `==`?**
> `is` compares identity (same object); `==` compares value equality.

**17. Why is `0.1 + 0.2 != 0.3`?**
> Floats use binary IEEE-754 and 0.1/0.2 can't be represented exactly. Compare with `math.isclose` or use `decimal`.

**18. What are the falsy values?**
> `0`, `0.0`, `0j`, `""`, `[]`, `()`, `{}`, `set()`, `range(0)`, `None`, `False`.

**19. What does `and`/`or` return?**
> One of the operands (not necessarily a bool): `0 or 5` → 5, `3 and 5` → 5. They short-circuit.

**20. Explain operator precedence with an example.**
> `2 + 3 * 4 ** 2 / 8`: `**` first (16), then `*` (48), `/` (6.0), then `+` → 8.0.

**21. What is type casting? Implicit vs explicit?**
> Converting between types. Implicit: Python does it (int + float → float). Explicit: you call `int()`, `float()`, `str()` …

**22. What does `int(3.99)` return? `round(2.5)`?**
> `3` (truncation). `2` (banker's rounding to nearest even).

**23. What is the walrus operator?**
> `:=` assigns inside an expression: `if (n := len(x)) > 3:`.

**24. What is `None`?**
> The single object of `NoneType`, representing "no value"; test with `is None`.

## 3. Strings
**25. Are strings mutable?**
> No. Methods return new strings.

**26. Explain slicing `s[a:b:c]`.**
> From index `a` up to but excluding `b` with step `c`; negatives count from the end; `s[::-1]` reverses.

**27. Difference between `find()` and `index()`?**
> Both return the first position; `find` returns −1 if missing, `index` raises `ValueError`.

**28. `split()` vs `partition()`?**
> `split` returns a list of all pieces; `partition` splits at the first separator into a 3-tuple `(before, sep, after)`.

**29. What is a raw string?**
> `r"..."` – backslashes are literal (useful for regex and Windows paths).

**30. Ways of formatting strings?**
> f-strings (`f"{x:.2f}"`), `str.format()`, `%` operator, `string.Template`.

**31. What is the difference between `str()` and `repr()`?**
> `str` is user-friendly; `repr` is unambiguous/developer-oriented (ideally valid code).

**32. How to check if a string is a palindrome?**
> `s == s[::-1]` (after cleaning case/punctuation).

**33. How does `join` work?**
> `sep.join(iterable_of_strings)` glues the items with `sep`; far faster than repeated `+=`.

**34. What are `ord()` and `chr()`?**
> Character ↔ Unicode code point: `ord("A") == 65`, `chr(65) == "A"`.

## 4. Collections
**35. List vs tuple?**
> List: mutable, `[]`; tuple: immutable, `()`, hashable (if contents are), slightly faster/smaller.

**36. Why is `(5)` not a tuple?**
> The comma makes a tuple: `(5,)`. Parentheses alone just group.

**37. What is a set? When to use it?**
> Unordered collection of unique hashable items; use for de-duplication, fast membership and set algebra.

**38. `{}` – set or dict?**
> An empty dict. Empty set is `set()`.

**39. Which types can be dict keys?**
> Hashable (immutable) objects: str, int, float, bool, tuple of hashables, frozenset. Not list/dict/set.

**40. Difference between `append()` and `extend()`?**
> `append(x)` adds x as a single element; `extend(it)` adds each element of `it`.

**41. `remove()` vs `pop()` vs `del`?**
> `remove(value)`: first matching value; `pop(i)`: removes and returns by index; `del l[i]`/`del l[a:b]`: delete by index/slice.

**42. `sort()` vs `sorted()`?**
> `sort` sorts in place and returns `None`; `sorted` returns a new list from any iterable.

**43. Shallow vs deep copy?**
> Shallow copies the container but shares nested objects; deep copy recursively duplicates everything (`copy.deepcopy`).

**44. What is a list comprehension?**
> `[expr for x in iterable if cond]` – concise list building.

**45. What does `dict.get(k, default)` do?**
> Returns `d[k]` or `default` (no `KeyError`).

**46. How to merge two dicts?**
> `{**a, **b}` or `a | b` (3.9+) or `a.update(b)`.

**47. How is a dictionary implemented?**
> A hash table with open addressing; average O(1) lookup, ordered by insertion since 3.7.

**48. How to remove duplicates from a list preserving order?**
> `list(dict.fromkeys(lst))`.

**49. How do you sort a dict by value?**
> `dict(sorted(d.items(), key=lambda kv: kv[1]))`.

**50. What are `Counter`, `defaultdict`, `deque`, `namedtuple`?**
> Counter counts items; defaultdict supplies a default for missing keys; deque is an O(1) double-ended queue; namedtuple gives tuple fields names.

## 5. Control flow
**51. `break` vs `continue` vs `pass`?**
> `break` exits the loop; `continue` skips to the next iteration; `pass` does nothing (placeholder).

**52. What is the `else` clause on a loop?**
> Runs when the loop ends without hitting `break`.

**53. `range(2, 10, 3)`?**
> 2, 5, 8 (stop excluded; lazy sequence).

**54. `for` vs `while`?**
> `for` iterates over an iterable (known items); `while` repeats while a condition is true (unknown count).

**55. Ternary in Python?**
> `x if condition else y`.

**56. What is `match-case`?**
> Structural pattern matching (3.10+) – a powerful switch that can destructure sequences, dicts and classes.

**57. What is an infinite loop? Example?**
> A loop that never ends: `while True:` without `break`. Useful for servers/menus with a break.

**58. What does `enumerate()` do?**
> Yields `(index, item)` pairs: `for i, x in enumerate(seq, start=1)`.

**59. What does `zip()` do?**
> Pairs items from several iterables, stopping at the shortest.

## 6. Functions
**60. Advantages of functions?**
> Reuse, modularity, readability, easier testing/debugging, abstraction.

**61. Types of arguments?**
> Positional, keyword, default, variable-length (`*args`, `**kwargs`), positional-only (`/`), keyword-only (`*`).

**62. `*args` and `**kwargs`?**
> `*args` collects extra positional arguments into a tuple; `**kwargs` collects extra keyword arguments into a dict.

**63. `return` vs `print`?**
> `return` sends a value back to the caller; `print` only displays it (function returns `None`).

**64. What is the LEGB rule?**
> Name lookup order: Local → Enclosing → Global → Built-in.

**65. `global` vs `nonlocal`?**
> `global x` refers to the module-level variable; `nonlocal x` refers to the nearest enclosing function's variable.

**66. What is a lambda?**
> An anonymous single-expression function: `lambda x: x * 2`.

**67. `map`, `filter`, `reduce`?**
> `map(f, it)` transforms each item; `filter(f, it)` keeps items where f is true; `functools.reduce(f, it)` folds to one value.

**68. Explain recursion. Base case?**
> A function calling itself on a smaller problem; the base case stops the recursion (otherwise `RecursionError`).

**69. Mutable default argument problem?**
> Defaults are evaluated once at definition, so a default list/dict is shared across calls. Use `None` and create inside.

**70. Are arguments passed by value or reference?**
> Pass by object reference ("pass by assignment"): mutating a mutable argument affects the caller; rebinding doesn't.

**71. What is a closure?**
> An inner function that remembers variables from its enclosing scope after the outer function returned.

**72. What is a decorator?**
> A function that takes a function and returns a modified one; `@dec` is sugar for `f = dec(f)`.

**73. What is a generator? `yield` vs `return`?**
> A function that produces values lazily; `yield` pauses and keeps state, `return` ends the function.

**74. Iterable vs iterator?**
> An iterable can produce an iterator (`__iter__`); an iterator yields items with `__next__` and raises `StopIteration` when exhausted.

**75. What is a docstring?**
> A string literal as the first statement of a module/class/function, available via `__doc__`/`help()`.

## 7. OOP
**76. Four pillars of OOP?**
> Encapsulation, abstraction, inheritance, polymorphism.

**77. Class vs object?**
> A class is a blueprint; an object is an instance created from it.

**78. What is `self`?**
> The instance on which a method is called, passed automatically as the first parameter.

**79. What is `__init__`?**
> The initializer called right after an object is created to set up its attributes.

**80. Instance vs class attributes?**
> Instance attributes belong to one object (`self.x`); class attributes are shared by all instances.

**81. Instance / class / static methods?**
> Instance: takes `self`. Class: `@classmethod`, takes `cls`, often alternative constructors. Static: `@staticmethod`, no implicit argument.

**82. How is encapsulation implemented?**
> Conventions `_protected`, `__private` (name mangling to `_Class__name`), and `@property` for controlled access.

**83. What is inheritance? Types?**
> A class acquiring another's attributes/methods. Single, multilevel, hierarchical, multiple, hybrid.

**84. What does `super()` do?**
> Returns a proxy to the next class in the MRO, used to call parent methods.

**85. What is method overriding? Overloading?**
> Overriding: subclass redefines a parent method. Overloading (same name, different signatures) isn't supported – use default args/`*args`/`singledispatch`.

**86. What is polymorphism?**
> One interface, many implementations; via overriding and duck typing.

**87. What is duck typing?**
> If an object has the needed methods/attributes it can be used, regardless of its class.

**88. What is MRO?**
> Method Resolution Order – the order classes are searched for a method (C3 linearization); see `Class.__mro__`.

**89. Abstract class?**
> A class with abstract methods (`abc.ABC`, `@abstractmethod`) that can't be instantiated until subclasses implement them.

**90. What are magic (dunder) methods? Examples.**
> Special methods called by Python syntax: `__init__`, `__str__`, `__repr__`, `__len__`, `__add__`, `__eq__`, `__getitem__`, `__call__`, `__enter__`/`__exit__`.

**91. `__str__` vs `__repr__`?**
> `__str__` for end users (`print`), `__repr__` for developers/debugging; if only `__repr__` exists it is used for both.

**92. What is operator overloading?**
> Defining dunder methods so operators work with your objects (`__add__` for `+`).

**93. Composition vs inheritance?**
> Composition = "has-a" (object contains others); inheritance = "is-a". Prefer composition for flexibility.

**94. What is `@property`?**
> Makes a method accessible like an attribute, enabling validation/computation behind attribute syntax.

**95. What is a dataclass?**
> `@dataclass` auto-generates `__init__`, `__repr__`, `__eq__` for classes that mainly hold data.

**96. What is a constructor? Can Python have multiple?**
> `__init__` is the initializer; only one per class, but use default args or `@classmethod` factories for alternatives.

**97. What is the diamond problem?**
> Ambiguity in multiple inheritance when a class inherits the same base through two paths; Python resolves it with C3 MRO.

## 8. Exceptions
**98. Errors vs exceptions?**
> Syntax errors are detected before running; exceptions are runtime errors that can be caught.

**99. `try / except / else / finally`?**
> `try` risky code; `except` handles; `else` runs if no exception; `finally` always runs (cleanup).

**100. Difference between `raise` and `assert`?**
> `raise` signals an exception deliberately; `assert cond, msg` checks a condition during development and raises `AssertionError` (removed with `-O`).

**101. How to create a custom exception?**
> `class MyError(Exception): pass` (optionally with extra fields and `super().__init__(msg)`).

**102. Why not use a bare `except:`?**
> It also catches `KeyboardInterrupt`/`SystemExit` and hides bugs; catch specific exceptions.

**103. What does `finally` do with `return`?**
> The `finally` block runs before the function actually returns; a `return` inside `finally` overrides earlier returns.

**104. What is exception chaining?**
> `raise New from old` records the original cause in `__cause__`.

**105. Common exceptions?**
> `ValueError`, `TypeError`, `ZeroDivisionError`, `IndexError`, `KeyError`, `NameError`, `AttributeError`, `FileNotFoundError`, `ImportError`.

## 9. Files
**106. File opening modes?**
> `r` read, `w` write (truncate), `a` append, `x` create-only, `+` update, `b` binary, `t` text.

**107. `read()`, `readline()`, `readlines()`?**
> Whole content as one string; one line; list of all lines.

**108. Why use `with open()`?**
> Guarantees the file is closed even if an error occurs (context manager).

**109. What do `seek()` and `tell()` do?**
> `tell()` reports the current position; `seek(offset, whence)` moves it.

**110. Text vs binary files?**
> Text files are decoded to `str` (with newline translation); binary files are raw `bytes` (images, PDFs).

**111. `pickle` vs `json`?**
> JSON: human-readable, language-independent, limited types. Pickle: Python-specific binary, handles most objects, unsafe to load from untrusted sources.

**112. How to read a CSV?**
> `csv.reader`/`csv.DictReader`, or `pandas.read_csv`.

## 10. Modules
**113. Module vs package vs library?**
> Module: one `.py` file. Package: folder with `__init__.py`. Library: collection of packages.

**114. Ways to import?**
> `import m`, `from m import x`, `import m as a`, `from m import *` (avoid).

**115. What does `if __name__ == "__main__":` do?**
> Runs code only when the file is executed directly, not when imported.

**116. `pip`? Virtual environment?**
> `pip` installs packages from PyPI. A virtual environment (`python -m venv`) isolates a project's dependencies.

**117. Useful standard modules?**
> `math`, `random`, `datetime`, `os`, `sys`, `pathlib`, `json`, `csv`, `re`, `collections`, `itertools`, `functools`, `logging`, `sqlite3`.

## 11. Advanced
**118. What is the GIL?**
> A lock in CPython allowing one thread to execute Python bytecode at a time; threads help I/O-bound work, processes help CPU-bound work.

**119. Threading vs multiprocessing vs asyncio?**
> Threads: shared memory, I/O concurrency. Processes: real parallelism, separate memory. asyncio: single-thread cooperative concurrency for many I/O tasks.

**120. How does memory management work?**
> Private heap; reference counting plus a cyclic garbage collector.

**121. What are type hints?**
> Optional annotations (`def f(x: int) -> str`) for tools and readers; not enforced at run time.

**122. What is a context manager?**
> An object with `__enter__`/`__exit__` used by `with` for setup/cleanup.

**123. What is a metaclass?**
> The class of a class (default `type`); controls how classes are created.

## 12. Algorithms & data structures
**124. Big-O of list `append`, `insert(0, x)`, `x in list`, `x in set`?**
> O(1) amortised, O(n), O(n), O(1) average.

**125. Binary search requirement and complexity?**
> Sorted data; O(log n).

**126. Which sort does Python use?**
> Timsort (hybrid merge + insertion), O(n log n) worst, O(n) best, stable.

**127. Stack vs queue?**
> Stack is LIFO (`append/pop`); queue is FIFO (`deque.append/popleft`).

**128. Array vs linked list?**
> Array/list: O(1) index, O(n) insert at front. Linked list: O(1) insert at known node, O(n) access.

**129. BFS vs DFS?**
> BFS explores level by level with a queue (shortest path in unweighted graphs); DFS goes deep first with a stack/recursion.

**130. What is memoisation / dynamic programming?**
> Caching sub-problem results to avoid recomputation (`functools.cache`).

## 13. Data libraries
**131. Why is NumPy faster than lists?**
> Contiguous typed memory and vectorised C loops.

**132. What is broadcasting?**
> Automatic stretching of smaller arrays to match shapes in element-wise operations.

**133. Series vs DataFrame?**
> Series: labelled 1-D array; DataFrame: labelled 2-D table.

**134. `loc` vs `iloc`?**
> `loc` selects by label; `iloc` by integer position.

**135. How to handle missing data in pandas?**
> `isna()`, `dropna()`, `fillna()`, interpolation.

## 14. Tricky "why" questions
**136. Why does `[[0]*3]*3` misbehave?**
> The outer `*` repeats the same inner list object three times; use a comprehension.

**137. Why does `x = y = []` share state?**
> Both names point to one list.

**138. Why is `a is b` sometimes True for small ints?**
> CPython caches ints −5…256 (an implementation detail; use `==`).

**139. Why can't you modify a list while iterating over it safely?**
> The iterator index shifts as items are removed, skipping elements.

**140. Why is string concatenation in a loop slow?**
> Each `+=` creates a new string (O(n²) overall); collect in a list and `join`.

**141. Why does `round(2.675, 2)` give 2.67?**
> 2.675 is stored as 2.67499999… in binary floating point.

**142. Why is `sum([0.1]*10) == 1.0` True in 3.12+ but a manual loop isn't?**
> Since 3.12 `sum()` uses compensated summation for floats.

**143. Why are tuples usable as dict keys but lists aren't?**
> Keys must be hashable; hash depends on contents, which lists can change.

**144. What happens if two modules import each other?**
> A circular import; the second import sees a partially initialised module. Fix by restructuring or importing inside functions.

**145. What is the difference between `deepcopy` and assignment?**
> Assignment just makes another name for the same object; deepcopy builds a new independent object graph.
