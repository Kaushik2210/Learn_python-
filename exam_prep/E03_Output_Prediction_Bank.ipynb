# 🔮 Exam Prep 03 — Output Prediction Bank (35 puzzles)

"What is the output?" is the most common question type in Python papers. The only way to get good is **to predict on paper first, then check**.

### The routine
1. Run the first code cell — it prints all the puzzles.
2. Pick puzzle **N**. Cover the screen. Trace it **by hand** (write variable values line by line).
3. Write your predicted output on paper.
4. Run `reveal(N)` in the answer cell. Compare **exactly** (spaces, quotes, newlines, `None`).
5. If wrong: figure out *why*, add it to your mistake log, redo it in 3 days.

> 🧠 The answers are produced by actually running the code, so they are always correct for the Python you have installed.

```python
import io, contextlib, textwrap

PUZZLES = [
# ---- Basics ----
("Operators", """
print(7 // 2, -7 // 2, 7 % -3, 2 ** 3 ** 2)
print(10 / 4, 10 // 4, 10 % 4, divmod(10, 4))
"""),
("Truthiness & logic", """
print(0 or 5, 3 and 0, None or "x", [] and 1, not "")
print(bool("False"), bool(0.0), 1 < 2 < 3, 1 < 3 < 2)
"""),
("Type surprises", """
print(type(3 / 1), type(3 // 1), type(True + 1), type([] == []))
print(int(3.99), int(-3.99), round(3.5), round(4.5), round(-0.5))
"""),
("Identity vs equality", """
a = [1, 2]; b = [1, 2]; c = a
print(a == b, a is b, a is c)
x = "hi"; y = "hi"
print(x == y)
"""),
# ---- Strings ----
("Slicing", """
s = "Programming"
print(s[3:8], s[-3:], s[::2], s[::-3], s[8:2:-2])
"""),
("String methods", """
s = "  Hello, World  "
print(s.strip().lower().split(", "), s.count("l"), s.find("o"), s.rstrip()[-3:])
"""),
("Formatting", """
x, y = 3.14159, 42
print(f"{x:.2f}|{y:5d}|{y:<5}|{y:05d}|{x:8.3f}|")
print("%d%%" % 50, "{:>6}".format("ab"))
"""),
("String multiplication & join", """
print("ab" * 3, "-".join("abc"), "a,b,,c".split(","), len("a\\nb"))
"""),
# ---- Lists ----
("List methods", """
L = [3, 1, 2]
L.append(5); L.insert(1, 9); L.remove(1)
print(L, L.pop(), L)
print(L.sort(), L)
"""),
("List aliasing", """
a = [1, 2, 3]
b = a
b.append(4)
c = a[:]
c.append(5)
print(a, b, c)
"""),
("List multiplication trap", """
grid = [[0] * 2] * 2
grid[0][0] = 7
print(grid)
"""),
("Comprehension", """
print([x * y for x in range(1, 4) for y in range(1, 3) if x != y])
print({x: x % 2 for x in range(4)}, [c for c in "hello" if c not in "aeiou"])
"""),
("Slicing assignment", """
L = list(range(8))
L[2:5] = ["a"]
print(L)
L[::2] = [0, 0, 0]
print(L)
"""),
# ---- Tuple/Set/Dict ----
("Sets", """
a = {1, 2, 3, 3, 2}
b = {3, 4}
print(a, a & b, a | b, a - b, a ^ b, len(a))
"""),
("Dict operations", """
d = {"a": 1, "b": 2}
d["c"] = d.get("c", 10) + 1
d.update({"a": 100})
print(d, d.pop("b"), list(d.items()), d.get("z", -1))
"""),
("Dict fromkeys trap", """
d = dict.fromkeys("ab", [])
d["a"].append(1)
print(d)
"""),
("Tuple", """
t = (1, [2, 3], 4)
t[1].append(9)
print(t, len(t), t[-1], t[1:])
a, *b, c = t
print(a, b, c)
"""),
# ---- Control flow ----
("Loops with else", """
for i in range(3):
    if i == 5:
        break
else:
    print("no break")
print(i)
"""),
("Nested loops", """
for i in range(1, 4):
    for j in range(i):
        print(i * j, end=" ")
    print()
"""),
("while with continue", """
n = 10
while n > 0:
    n -= 3
    if n % 2 == 0:
        continue
    print(n, end=" ")
print("| end", n)
"""),
("Range", """
print(list(range(5, 0, -2)), list(range(3, 3)), sum(range(1, 5)), len(range(0, 10, 3)))
"""),
# ---- Functions ----
("Default argument trap", """
def f(x, acc=[]):
    acc.append(x)
    return acc
print(f(1), f(2), f(3, []))
"""),
("Scope", """
x = 10
def f():
    x = 20
    def g():
        nonlocal x
        x += 1
        return x
    return g()
print(f(), x)
"""),
("args / kwargs", """
def f(a, b=2, *args, **kw):
    return a, b, args, kw
print(f(1), f(1, 3, 5, 7, z=9))
"""),
("Lambda / map / filter", """
print(list(map(lambda x: x ** 2, filter(lambda x: x % 2, range(6)))))
print((lambda a, b=3: a * b)(2))
"""),
("Recursion order", """
def f(n):
    if n == 0:
        return
    print(n, end=" ")
    f(n - 1)
    print(n, end=" ")
f(3)
"""),
("Closure gotcha", """
fs = [lambda: i for i in range(3)]
print([f() for f in fs])
gs = [lambda i=i: i for i in range(3)]
print([g() for g in gs])
"""),
("Generator", """
def g():
    print("start")
    yield 1
    print("middle")
    yield 2
    print("end")
it = g()
print("created")
print(next(it))
print(next(it))
"""),
# ---- OOP ----
("Class vs instance attributes", """
class A:
    x = 1
    def __init__(self): self.y = 2
a1, a2 = A(), A()
a1.x = 10
A.x = 100
print(a1.x, a2.x, A.x, a1.y)
"""),
("Inheritance & super", """
class P:
    def hello(self): return "P"
class C(P):
    def hello(self): return "C>" + super().hello()
class D(C): pass
print(D().hello(), isinstance(D(), P), issubclass(P, D))
"""),
("__str__ vs __repr__", """
class T:
    def __str__(self): return "str!"
    def __repr__(self): return "repr!"
t = T()
print(t, [t], str(t), repr(t), f"{t}")
"""),
("Method resolution order", """
class A:
    def go(self): return "A"
class B(A):
    def go(self): return "B" + super().go()
class C(A):
    def go(self): return "C" + super().go()
class D(B, C):
    def go(self): return "D" + super().go()
print(D().go(), [k.__name__ for k in D.__mro__])
"""),
# ---- Exceptions ----
("try/except/else/finally", """
def f(x):
    try:
        print("A", end=" ")
        r = 10 / x
    except ZeroDivisionError:
        print("B", end=" ")
        return "err"
    else:
        print("C", end=" ")
        return r
    finally:
        print("D", end=" ")
print(f(5))
print(f(0))
"""),
("finally overrides", """
def f():
    try:
        return "try"
    finally:
        return "finally"
print(f())
"""),
("Exception order", """
try:
    try:
        int("x")
    except TypeError:
        print("inner TypeError")
    finally:
        print("inner finally")
except ValueError as e:
    print("outer ValueError:", e)
"""),
]

def show_puzzle(n):
    title, code = PUZZLES[n - 1]
    print(f"━━━━━━━━ Puzzle {n}: {title} ━━━━━━━━")
    print(textwrap.dedent(code).strip())
    print()

for n in range(1, 13):
    show_puzzle(n)
```

## 👀 Reveal an answer (only after you predicted!)
Change the number and run the next cell. Leave it un-run until you're ready.

```python
# notest
def reveal(n):
    title, code = PUZZLES[n - 1]
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(textwrap.dedent(code), {})
    print(f"Puzzle {n}: {title}\n{'-' * 30}")
    print(buf.getvalue())

# reveal(1)     # <- remove the leading # and run when you have predicted puzzle 1
```

## 🧾 Explanations (read AFTER you have revealed)

| # | Key idea |
|---|---|
| 1 | `//` floors toward −∞, `%` takes the sign of the divisor, `**` is right-associative (`2 ** 9`) |
| 2 | `and`/`or` return an operand; chained comparisons `1 < 3 < 2` mean `(1<3) and (3<2)` |
| 3 | `int()` truncates toward 0; `round` uses banker's rounding (ties → even) |
| 4 | `==` compares values, `is` compares identity; equal strings **may** be the same object (never rely on it) |
| 5 | Slice `[start:stop:step]`, stop excluded; negative steps walk backwards |
| 6 | Methods return new strings; `count`, `find`; `strip` before `lower` |
| 7 | Format mini-language: `.2f` decimals, `5d` width, `<` left-align, `05d` zero-pad |
| 8 | `split(",")` keeps empty strings; `len("a\nb")` is 3 (escape is one char) |
| 9 | `sort()` and `append()` return `None`; `pop()` removes last. Note that `print(L, L.pop(), L)` evaluates arguments left-to-right, but both `L`s are the same object printed after `pop` ran |
| 10 | `b = a` aliases; `a[:]` copies |
| 11 | `[[0]*2]*2` repeats one row object |
| 12 | Nested comprehension order = nested `for` order; `if` filters |
| 13 | Slice assignment can shrink/grow; extended slices need equal length |
| 14 | Set algebra; duplicates disappear |
| 15 | `get` with default, `update`, `pop` returns value — again argument evaluation order matters |
| 16 | `fromkeys` shares the single list object |
| 17 | Tuple is immutable but its list element isn't; star-unpacking |
| 18 | Loop `else` runs unless `break`; loop variable persists |
| 19 | Inner loop restarts each outer iteration |
| 20 | `continue` skips the print |
| 21 | `range` semantics; empty range; `len(range(0,10,3))` = 4 |
| 22 | Mutable default shared; passing your own list avoids it |
| 23 | `nonlocal` modifies the enclosing `x` (20 → 21); global `x` untouched |
| 24 | `*args` tuple, `**kw` dict |
| 25 | `map`/`filter` are lazy; lambda default arg |
| 26 | Print-before-recursion vs print-after-recursion |
| 27 | Late binding closures vs default-arg capture |
| 28 | Generators run only when iterated; output interleaves |
| 29 | Instance attribute shadows class attribute |
| 30 | `super()` follows the MRO |
| 31 | `print` uses `__str__`; containers use `__repr__` |
| 32 | `super()` in a diamond visits each class once in MRO order |
| 33 | `else` only without exception; `finally` always |
| 34 | `return` in `finally` wins |
| 35 | Exception in `try` → `finally` → propagates to the matching outer handler |
