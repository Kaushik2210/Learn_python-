````python
# --- Paper 1, Q13: try / except / else / finally with input 0 ---
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Result:", result)
    finally:
        print("Finished")

divide(10, 0)      # Cannot divide by zero / Finished
divide(10, 4)      # Result: 2.5 / Finished

# --- Paper 1, Q14: single inheritance with super() ---
class Person:
    def __init__(self, name):
        self.name = name
    def intro(self):
        return f"I am {self.name}"

class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll
    def intro(self):
        return super().intro() + f", roll {self.roll}"

print(Student("Asha", 7).intro())

# --- Paper 1, Q15 ---
s = "Programming"
print(s[::-1], s[2:5], s[-3:])
````

````python
# --- Paper 1, Q16: primes ---
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    return True

print("Primes 1-50:", [n for n in range(1, 51) if is_prime(n)])

# --- Paper 1, Q17: sentence statistics ---
def sentence_stats(text):
    words = [w.strip(".,!?;:").lower() for w in text.split()]
    vowels = sum(1 for ch in text.lower() if ch in "aeiou")
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    most = max(freq, key=freq.get)
    return len(words), vowels, most

print(sentence_stats("The cat and the dog saw the bird."))

# --- Paper 1, Q18: BankAccount ---
class InsufficientFunds(Exception):
    pass

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner, self.balance = owner, balance
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.balance += amount
    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFunds(f"Balance {self.balance} < {amount}")
        self.balance -= amount
    def __str__(self):
        return f"{self.owner}: Rs.{self.balance}"

acc = BankAccount("Asha", 500)
acc.deposit(250)
acc.withdraw(100)
print(acc)
try:
    acc.withdraw(10_000)
except InsufficientFunds as e:
    print("Error:", e)

# --- Paper 1, Q19: file processing ---
with open("input.txt", "w") as f:
    f.write("python is easy\nfiles are useful\n")

lines = words_total = 0
upper_words = []
with open("input.txt") as src:
    for line in src:
        lines += 1
        ws = line.split()
        words_total += len(ws)
        upper_words += [w.upper() for w in ws]
with open("output.txt", "w") as dst:
    dst.write("\n".join(upper_words))
print(lines, words_total, open("output.txt").read().split())

# --- Paper 1, Q20: bubble sort + binary search ---
def bubble_sort(a):
    a = a[:]
    for i in range(len(a) - 1):
        swapped = False
        for j in range(len(a) - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        if not swapped:
            break
    return a

def binary_search(a, target):
    lo, hi = 0, len(a) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if a[mid] == target:
            return mid
        lo, hi = (mid + 1, hi) if a[mid] < target else (lo, mid - 1)
    return -1

data = [64, 25, 12, 22, 11, 90]
sorted_data = bubble_sort(data)
print(sorted_data, binary_search(sorted_data, 25), binary_search(sorted_data, 99))
````

````python
# --- Paper 2, Q11: shallow vs deep copy ---
import copy
original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)
original[0][0] = 99
print("original:", original)
print("shallow :", shallow, "(shares inner lists)")
print("deep    :", deep, "(independent)")

# --- Paper 2, Q12: LEGB, global, nonlocal ---
counter = 0
def bump():
    global counter
    counter += 1

def make_counter():
    n = 0
    def inc():
        nonlocal n
        n += 1
        return n
    return inc

bump(); bump()
c = make_counter()
print(counter, c(), c())

# --- Paper 2, Q15 ---
def gen(n):
    for i in range(n):
        yield i * i
print(list(gen(5)), sum(gen(4)))
````

````python
# --- Paper 2, Q16: Fibonacci, loop and recursion ---
def fib_loop(n):
    a, b, out = 0, 1, []
    for _ in range(n):
        out.append(a)
        a, b = b, a + b
    return out

def fib_rec(n):
    return n if n < 2 else fib_rec(n - 1) + fib_rec(n - 2)

print(fib_loop(10), [fib_rec(i) for i in range(10)])

# --- Paper 2, Q17: marks, percentage, grade ---
def total(marks):
    return sum(marks)

def percentage(marks, max_per_subject=100):
    return total(marks) / (len(marks) * max_per_subject) * 100

def grade(pct):
    if pct >= 90: return "A+"
    if pct >= 80: return "A"
    if pct >= 70: return "B"
    if pct >= 60: return "C"
    if pct >= 40: return "D"
    return "F"

marks = [78, 85, 92, 66, 74]          # (input() in the real exam: list(map(int, input().split())))
pct = percentage(marks)
print(f"Total={total(marks)}  Percentage={pct:.2f}%  Grade={grade(pct)}")

# --- Paper 2, Q18: polymorphism with an abstract class ---
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self): ...

class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return math.pi * self.r ** 2

class Rectangle(Shape):
    def __init__(self, w, h): self.w, self.h = w, h
    def area(self): return self.w * self.h

for shape in [Circle(2), Rectangle(3, 4), Circle(1)]:
    print(f"{type(shape).__name__:10} area = {shape.area():.2f}")

# --- Paper 2, Q19: CSV -> topper, average, JSON of passed students ---
import csv, json
with open("class.csv", "w", newline="") as f:
    csv.writer(f).writerows([["name", "marks"], ["Asha", 88], ["Ravi", 35], ["Meena", 72], ["Tom", 40]])

with open("class.csv", newline="") as f:
    rows = [{"name": r["name"], "marks": int(r["marks"])} for r in csv.DictReader(f)]
topper = max(rows, key=lambda r: r["marks"])
average = sum(r["marks"] for r in rows) / len(rows)
passed = [r for r in rows if r["marks"] >= 40]
with open("passed.json", "w") as f:
    json.dump(passed, f, indent=2)
print("Topper:", topper["name"], "| Average:", average, "| Passed:", [r["name"] for r in passed])

# --- Paper 2, Q20: Stack + balanced brackets ---
class Stack:
    def __init__(self): self.items = []
    def push(self, x): self.items.append(x)
    def pop(self): return self.items.pop()
    def peek(self): return self.items[-1]
    def is_empty(self): return not self.items

def balanced(expr):
    pairs = {")": "(", "]": "[", "}": "{"}
    st = Stack()
    for ch in expr:
        if ch in "([{":
            st.push(ch)
        elif ch in pairs:
            if st.is_empty() or st.pop() != pairs[ch]:
                return False
    return st.is_empty()

print(balanced("{[(a+b)*c]}"), balanced("(a+b]"), balanced("((a)"))
````
