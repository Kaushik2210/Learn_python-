## 1. What is Python?

**Python** is a *high-level, interpreted, general-purpose, dynamically-typed* programming language created by **Guido van Rossum** and first released in **1991**. The name comes from the comedy group *Monty Python*, not the snake.

Let's unpack those buzzwords (examiners LOVE these):

| Term | Meaning |
|---|---|
| **High-level** | Reads close to English; you don't manage memory or hardware details |
| **Interpreted** | Code is executed line-by-line by an *interpreter* (no separate compile-then-run step visible to you) |
| **General-purpose** | Web, data science, AI, automation, games, scripting, embedded... |
| **Dynamically typed** | You never declare a variable's type; the type belongs to the *value*, checked at run time |
| **Object-oriented** | Everything is an object (numbers, strings, functions, even classes) |
| **Multi-paradigm** | Supports procedural, object-oriented and functional styles |
| **Garbage collected** | Python frees unused memory automatically |

### ✅ Key features of Python (memorise this list for exams)
- Easy to learn, read and write (clean syntax)
- Free and open source
- Portable / cross-platform (Windows, macOS, Linux)
- Huge standard library ("batteries included") + 500k+ third-party packages on PyPI
- Supports OOP, functional and procedural programming
- Dynamically typed, automatic memory management
- Extensible & embeddable (can call C/C++, can be embedded in apps)
- Large community

### ⚠️ Disadvantages (also asked!)
- Slower than compiled languages like C/C++ (interpreted + dynamic)
- Higher memory consumption
- Weak for mobile app development
- Runtime errors show up late (dynamic typing)
- The GIL (Global Interpreter Lock) limits true multi-threaded CPU parallelism in CPython

## 2. How does Python run your code?

```
 your_code.py  ──►  Compiler  ──►  Bytecode (.pyc)  ──►  Python Virtual Machine (PVM)  ──►  Output
   (source)        (automatic)    (intermediate)          (interprets the bytecode)
```

1. **Source code** (`.py`) is what you write.
2. Python **compiles** it (automatically, invisibly) to **bytecode** – a low-level, platform-independent form. Cached in `__pycache__/*.pyc`.
3. The **PVM** (Python Virtual Machine) executes the bytecode one instruction at a time.

So Python is *both* compiled and interpreted — the "compile" step is just hidden. That's a classic trick exam question.

### Interpreter vs Compiler

| | **Compiler** (C, C++) | **Interpreter** (Python) |
|---|---|---|
| Translates | Whole program at once | Line by line / bytecode instruction by instruction |
| Output | Machine-code executable file | No standalone executable (runs via the interpreter) |
| Error reporting | All syntax errors listed after compiling | Stops at the first error it hits |
| Speed | Faster execution | Slower execution |
| Debugging | Harder | Easier |

### Implementations of Python
- **CPython** – the standard one, written in C (what you have installed)
- **PyPy** – faster, has a JIT compiler
- **Jython** – runs on the Java VM
- **IronPython** – runs on .NET

````python
import dis

def add(a, b):
    return a + b

dis.dis(add)
````

## 3. Your first program: `print()`

`print()` is a **built-in function** that displays output on the screen. Whatever you put inside the parentheses is shown.

````python
print("Hello, World!")
````

Things to notice:
- Text (a **string**) goes inside quotes. Single `'...'` or double `"..."` both work.
- `print` is all lowercase. `Print("hi")` would be a `NameError` – Python is **case-sensitive**.

### Printing several things

`print()` can take many arguments separated by commas. By default it puts a **space** between them and a **newline** at the end.

````python
print("Python", "is", "fun")
print(2024, 3.14, True)      # non-strings are converted automatically
print("Total:", 10 + 5)      # expressions are evaluated first
````

### `sep` and `end` — control the formatting

- `sep` = what goes **between** items (default `" "`)
- `end` = what goes **after** the last item (default `"\n"`)

````python
print("2025", "01", "15", sep="-")       # 2025-01-15
print("A", "B", "C", sep="")             # ABC
print("Loading", end="...")              # stays on the same line
print("done!")
print("a", "b", sep="\n")                # each item on its own line
````

### Escape sequences

A backslash `\` inside a string starts a special character.

| Sequence | Meaning |
|---|---|
| `\n` | new line |
| `\t` | tab |
| `\\` | a literal backslash |
| `\'` `\"` | a literal quote |
| `\r` | carriage return |

````python
print("Line 1\nLine 2")
print("Name\tAge\nRam\t20")
print("She said \"hi\"")
print('It\'s fine')
print("C:\\Users\\Kaushik")
print(r"C:\Users\new")     # raw string: backslashes are NOT special
````

## 4. Comments — notes for humans

Python ignores anything after `#`. Use comments to explain **why**, not what.

For multi-line explanations people use **docstrings** (triple-quoted strings) — they're technically strings, but placed at the start of a module/function/class they become official documentation.

````python
# This is a single-line comment
print("Comments are ignored")   # inline comment

"""
This is a triple-quoted string. Often used as a multi-line comment,
but its real job is docstrings (documentation).
"""

def greet():
    """Return a friendly greeting. (This is a docstring.)"""
    return "hi"

print(greet.__doc__)
````

## 5. Indentation — Python's most important rule

Other languages use `{ }` to group code. Python uses **indentation** (spaces at the start of a line). The convention is **4 spaces** per level. All lines in the same block must have the same indentation.

````python
age = 20
if age >= 18:
    print("Adult")          # inside the if-block (4 spaces)
    print("Can vote")       # same block
print("This always prints") # outside the block
````

What happens with wrong indentation? `IndentationError`. Let's provoke it safely using `exec` so the notebook doesn't crash:

````python
bad_code = """
if True:
print("oops")
"""
try:
    exec(bad_code)
except IndentationError as e:
    print("IndentationError:", e)
````

## 6. Statements, lines and multi-line code

- One statement per line normally.
- Put several on one line with `;` (legal but **discouraged**).
- Continue a long statement on the next line with `\` **or** inside `()`, `[]`, `{}` (preferred).

````python
a = 1; b = 2; print(a + b)      # discouraged style

total = 1 + 2 + 3 + \
        4 + 5                   # backslash continuation
print(total)

total2 = (1 + 2 + 3 +
          4 + 5)                # parentheses continuation (preferred)
print(total2)
````

## 8. Errors — the three kinds (exam favourite!)

| Type | When it happens | Example |
|---|---|---|
| **Syntax error** | Code breaks Python's grammar; caught *before* running | missing colon, unclosed bracket |
| **Runtime error (exception)** | Code is valid but fails while running | dividing by zero, wrong type, missing file |
| **Logical error** | Program runs but gives the **wrong answer** | using `+` where you meant `*` |

Logical errors are the worst — Python can't warn you because it does exactly what you told it to do.

````python
# 1) Syntax error (we use compile() to show it without crashing)
try:
    compile("print('hi'", "<demo>", "exec")
except SyntaxError as e:
    print("SyntaxError:", e.msg)

# 2) Runtime error
try:
    print(10 / 0)
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)

# 3) Logical error: intended "average of 4 and 6" but forgot parentheses
avg = 4 + 6 / 2
print("Wrong average:", avg, "  Correct:", (4 + 6) / 2)
````

### How to read a traceback
When an error is *not* handled, Python prints a **traceback**. Read it **from the bottom up**:

```
Traceback (most recent call last):
  File "demo.py", line 5, in <module>      ← where it happened (top = outermost call)
    result = divide(10, 0)
  File "demo.py", line 2, in divide
    return a / b
ZeroDivisionError: division by zero         ← THE ACTUAL ERROR: read this first!
```

1. Read the **last line** – error type + message.
2. Look at the line above – the exact line that failed.
3. Work upward through the calls to see how you got there.

## 9. Python style: PEP 8 and the Zen of Python

**PEP 8** is the official style guide. Highlights:
- 4 spaces per indent, never mix tabs and spaces
- `snake_case` for variables/functions, `PascalCase` for classes, `UPPER_CASE` for constants
- Max ~79 characters per line
- Spaces around operators: `x = a + b` (not `x=a+b`)
- Two blank lines between top-level functions

The **Zen of Python** is Python's philosophy. Run this famous easter egg:

````python
import this
````

## 10. Keywords & identifiers

**Identifiers** are names you give to variables, functions, classes. Rules:
1. Start with a letter or underscore (`_`) — **not** a digit
2. Contain only letters, digits, underscores
3. Case-sensitive (`name` ≠ `Name`)
4. Cannot be a **keyword**

**Keywords** are reserved words with special meaning. Python has ~35 of them:

````python
import keyword
print(len(keyword.kwlist), "keywords:")
print(keyword.kwlist)
````

````python
# valid vs invalid identifiers
valid   = ["age", "_hidden", "student_name", "x1", "Total", "__init__"]
invalid = ["1st", "my-name", "class", "a b", "@home"]

for name in valid + invalid:
    ok = name.isidentifier() and not keyword.iskeyword(name)
    print(f"{name!r:15} -> {'valid' if ok else 'INVALID'}")
````

# 🕵️ Hidden Facts & Pro Tricks (Module 01)

**Fun fact 1:** Python began as a Christmas-holiday hobby project in December 1989. Guido van Rossum named it after *Monty Python's Flying Circus*.

**Fun fact 2:** `import this` (the Zen of Python) and `import antigravity` (opens an xkcd comic in your browser) are official easter eggs. `from __future__ import braces` is a joke too — try it:

````python
try:
    exec("from __future__ import braces")
except SyntaxError as e:
    print("Python says:", e)
````

**Trick 1 — The `_` variable:** in the REPL, `_` holds the result of the last expression.

**Trick 2 — `print` can write to files:** `print("hi", file=open("out.txt", "w"))`.

**Trick 3 — Debug with `=` in f-strings:** `print(f"{x=}")` shows the name AND the value.

**Trick 4 — `python -m http.server`** instantly serves the current folder at http://localhost:8000 (great for sharing files on Wi‑Fi).

**Trick 5 — `python -i script.py`** runs a script and drops you into the REPL with all its variables — superb for debugging.

**Trick 6 — `python -m timeit "sum(range(1000))"`** benchmarks a snippet with no code.

**Trick 7 — `breakpoint()`** (3.7+) drops into the debugger at that line.

**Trick 8 — `python -c "print(2**100)"`** runs one-liners from the shell.

````python
x, y = 3, "py"
print(f"{x=}  {y=}  {x*2=}")

import sys
print(sys.version_info[:3], sys.platform)
print(sys.getsizeof(0), sys.getsizeof(2**100), "bytes  <- ints grow as needed")
````

**Hidden fact — Python isn't slow because it is 'interpreted'; it's slow because everything is a flexible object.** Even `5` is a full object with a type pointer and reference count. That's why NumPy (C arrays) is 50–100× faster for number crunching.

**Hidden fact — Python 3.11+** got roughly 25% faster than 3.10, and 3.13 ships an experimental JIT and an optional *free-threaded* (no-GIL) build.
