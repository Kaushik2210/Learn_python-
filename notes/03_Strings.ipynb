## 1. Creating strings

A string (`str`) is an **immutable sequence of Unicode characters**.

````python
s1 = 'single quotes'
s2 = "double quotes"
s3 = "It's easy to include an apostrophe"
s4 = 'He said "wow"'
s5 = """Triple quotes
can span
multiple lines"""
s6 = '''Also works
with single quotes'''
empty = ""

print(s3)
print(s5)
print(type(s1), len(s5), repr(empty))
````

`str()` converts anything to a string. `len()` returns the number of characters.

````python
print(str(123) + "4", len("Hello, World!"), str([1, 2]), str(None))

# Concatenation (+) and repetition (*)
print("Py" + "thon", "-" * 20, "ab" * 3)
# print("Age: " + 20)         # TypeError! must convert:
print("Age: " + str(20))
````

## 2. Indexing & slicing

Every character has a **position (index)**. Forward indexes start at **0**; backward indexes start at **-1**.

```
 P   y   t   h   o   n
 0   1   2   3   4   5     ← positive index
-6  -5  -4  -3  -2  -1     ← negative index
```

````python
s = "Python"
print(s[0], s[3], s[-1], s[-6])
try:
    print(s[10])
except IndexError as e:
    print("IndexError:", e)
````

### Slicing: `s[start : stop : step]`
- includes `start`, **excludes** `stop`
- defaults: start=0, stop=len, step=1
- **slicing never raises IndexError** (out-of-range is clipped)

````python
s = "Python Programming"
print(s[0:6])       # 'Python'
print(s[:6])        # same; start omitted
print(s[7:])        # 'Programming'
print(s[-11:])      # last 11 chars
print(s[::2])       # every 2nd char
print(s[::-1])      # REVERSED string!
print(s[6:0:-1])    # backwards from index 6 to 1
print(s[2:100])     # out of range is fine
print(s[5:2])       # '' (start > stop with positive step)
print(s[:])         # a full copy
````

### Visualising the slice `s[a:b]`
Think of the indexes as pointing to the **gaps between** characters:

```
 | P | y | t | h | o | n |
 0   1   2   3   4   5   6
```
`s[1:4]` → `'yth'` (gap 1 to gap 4).

## 3. Strings are immutable
You can't modify a character in place. You build a *new* string.

````python
s = "hello"
try:
    s[0] = "J"
except TypeError as e:
    print("TypeError:", e)

s = "J" + s[1:]        # build a new string
print(s)
print(s.replace("l", "L"))   # methods return NEW strings; original is unchanged
print(s)
````

## 4. String methods (the complete toolbox)

> All methods return a **new** string (or another value); the original never changes.

### 4.1 Case conversion

````python
t = "hello WORLD, welcome to pYthon"
print(t.upper())
print(t.lower())
print(t.title())          # Each Word Capitalised
print(t.capitalize())     # only the first char
print(t.swapcase())
print("ß".casefold())     # aggressive lowercase for caseless comparison
````

### 4.2 Searching & counting

| Method | Returns |
|---|---|
| `find(sub)` | lowest index or **-1** if missing |
| `rfind(sub)` | highest index or -1 |
| `index(sub)` | like `find` but raises **ValueError** |
| `count(sub)` | number of non-overlapping occurrences |
| `startswith(p)` / `endswith(p)` | bool |

````python
s = "banana bandana"
print(s.find("an"), s.rfind("an"), s.find("xyz"))
print(s.count("an"), s.count("a"))
try:
    s.index("xyz")
except ValueError as e:
    print("ValueError:", e)
print(s.startswith("ban"), s.endswith("ana"))
print(s.find("an", 3))        # search starting from index 3
print(s.startswith(("ban", "xyz")))   # tuple of prefixes
````

### 4.3 Testing content (all return `bool`)

| Method | True if… |
|---|---|
| `isalpha()` | all letters (and non-empty) |
| `isdigit()` / `isnumeric()` / `isdecimal()` | all digits |
| `isalnum()` | letters or digits only |
| `isspace()` | only whitespace |
| `isupper()` / `islower()` | cased letters all upper/lower |
| `istitle()` | Title Case |
| `isidentifier()` | valid variable name |

````python
tests = ["abc", "123", "abc123", "  ", "ABC", "Hello World", "a b", ""]
for t in tests:
    print(f"{t!r:15} alpha={t.isalpha()!s:5} digit={t.isdigit()!s:5} alnum={t.isalnum()!s:5} "
          f"space={t.isspace()!s:5} upper={t.isupper()!s:5}")
````

### 4.4 Stripping, padding & aligning

````python
s = "   padded text \n"
print(repr(s.strip()))       # both ends
print(repr(s.lstrip()))
print(repr(s.rstrip()))
print("xxhixx".strip("x"))   # strip specific characters

print("42".zfill(6))         # 000042
print("hi".center(11, "*"))
print("hi".ljust(8, ".") + "|")
print("hi".rjust(8, ".") + "|")
````

### 4.5 Splitting & joining ⭐ (VERY important)

- `str.split(sep=None, maxsplit=-1)` → list of pieces
- `sep.join(iterable)` → glue pieces with `sep`
- `partition(sep)` → 3-tuple `(before, sep, after)`
- `splitlines()` → split at line breaks

````python
line = "apple,banana,cherry,date"
fruits = line.split(",")
print(fruits)
print(line.split(",", 2))          # at most 2 splits
print("  many   spaces  here ".split())   # no arg: split on any whitespace & drop empties

print("-".join(fruits))
print(" ".join(["Python", "is", "cool"]))
print("".join(["a", "b", "c"]))
print("email@site.com".partition("@"))
print("a.b.c".rpartition("."))
print("line1\nline2\r\nline3".splitlines())

# join needs strings!
nums = [1, 2, 3]
print(",".join(str(n) for n in nums))
````

### 4.6 Replacing & translating

````python
s = "I like cats. Cats are cute."
print(s.replace("cats", "dogs"))         # case-sensitive
print(s.replace("Cats", "Dogs", 1))      # only first occurrence
table = str.maketrans("aeiou", "12345")
print("education".translate(table))
print("hello world".translate(str.maketrans("", "", "lo")))   # delete chars l and o
````

## 5. String formatting

### 5.1 f-strings (Python 3.6+) — **use these!**

````python
name, age, pi = "Asha", 19, 3.14159265

print(f"My name is {name} and I am {age}.")
print(f"Next year: {age + 1}")
print(f"Upper: {name.upper()}")
print(f"{pi:.2f}")                # 2 decimal places
print(f"{age=}")                  # debugging: prints  age=19
print(f"{name!r}")                # repr()
print(f"{{literal braces}}")
````

### 5.2 Format specification mini-language: `{value:fill align sign width ,.precision type}`

````python
n = 1234567.891
print(f"{n:,.2f}")          # thousands separator: 1,234,567.89
print(f"{n:15.1f}|")        # width 15
print(f"{'left':<10}|{'mid':^10}|{'right':>10}|")
print(f"{'pad':*^11}")
print(f"{42:05d}")          # 00042
print(f"{42:+d}")           # +42
print(f"{255:b} {255:o} {255:x} {255:X} {255:#x}")   # bases
print(f"{0.256:.1%}")       # 25.6%
print(f"{12345.678:e}")     # scientific
print(f"{5:>3}|{50:>3}|{500:>3}|")

# Table example
items = [("Pen", 10, 2.5), ("Notebook", 5, 45.0), ("Bag", 1, 799.99)]
print(f"{'Item':<10}{'Qty':>5}{'Price':>10}")
print("-" * 25)
for item, qty, price in items:
    print(f"{item:<10}{qty:>5}{price:>10.2f}")
````

### 5.3 `str.format()` and `%` (older styles — appear in exams & legacy code)

````python
print("Hello, {}! You are {}.".format("Ravi", 21))
print("{0} {1} {0}".format("a", "b"))
print("{name} scored {marks:.1f}".format(name="Ravi", marks=87.456))

print("Name: %s, Age: %d, Height: %.2f" % ("Ravi", 21, 5.9))
print("%5d|%-5d|%05d" % (42, 42, 42))
print("%x %o %e %%" % (255, 8, 1234.5))
````

## 7. Characters & Unicode

Every character has a numeric **code point**. `ord()` → number, `chr()` → character.

````python
print(ord("A"), ord("a"), ord("0"), ord(" "))
print(chr(72), chr(105), chr(8364))
print(chr(ord("a") + 1))                # next letter -> 'b'

# Caesar cipher shifting by 3 (lowercase only)
word = "xyz"
enc = "".join(chr((ord(c) - 97 + 3) % 26 + 97) for c in word)
print(enc)

# Encoding text to bytes and back
b = "héllo".encode("utf-8")
print(b, len("héllo"), len(b))
print(b.decode("utf-8"))
````

## 10. Classic string programs (know these cold!)

````python
# 1. Reverse a string
def reverse(s):
    return s[::-1]
print(reverse("python"))

# 2. Palindrome check (ignore case/spaces/punctuation)
def is_palindrome(s):
    cleaned = "".join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]
print(is_palindrome("A man, a plan, a canal: Panama"), is_palindrome("hello"))

# 3. Count vowels
def count_vowels(s):
    return sum(1 for c in s.lower() if c in "aeiou")
print(count_vowels("Programming Language"))

# 4. Character frequency
def frequency(s):
    freq = {}
    for c in s:
        freq[c] = freq.get(c, 0) + 1
    return freq
print(frequency("mississippi"))

# 5. Anagram check
def is_anagram(a, b):
    return sorted(a.lower()) == sorted(b.lower())
print(is_anagram("Listen", "Silent"), is_anagram("abc", "abd"))

# 6. Word count / longest word
sentence = "the quick brown fox jumps over the lazy dog"
words = sentence.split()
print(len(words), max(words, key=len))

# 7. Toggle case, capitalise every word
print("hello world".swapcase(), " ".join(w.capitalize() for w in "hello big world".split()))

# 8. Remove duplicates keeping order
print("".join(dict.fromkeys("programming")))

# 9. Check if string is a valid identifier-like username
def valid_username(u):
    return 3 <= len(u) <= 12 and u[0].isalpha() and u.isalnum()
print(valid_username("kaushik22"), valid_username("2fast"))

# 10. Compress: aaabbc -> a3b2c1
def compress(s):
    if not s:
        return ""
    out, count = [], 1
    for prev, cur in zip(s, s[1:]):
        if cur == prev:
            count += 1
        else:
            out.append(f"{prev}{count}")
            count = 1
    out.append(f"{s[-1]}{count}")
    return "".join(out)
print(compress("aaabbc"))
````

## 11. Performance tip: building strings

Because strings are immutable, `s += "x"` in a big loop creates a new string each time. Collect pieces in a **list** and `"".join()` once.

````python
pieces = []
for i in range(5):
    pieces.append(str(i))
result = "".join(pieces)
print(result)
````

# 🕵️ Hidden Facts & Pro Tricks (Module 03)

````python
# 1. Adjacent string literals are joined automatically (great for long strings)
msg = ("Hello, "
       "this is one string "
       "built from three literals")
print(msg)

# 2. Slice objects can be reused
first3 = slice(0, 3)
print("python"[first3], [10, 20, 30, 40][first3])

# 3. Quick banners
print("=" * 30)
print("MENU".center(30, "="))

# 4. Reverse words vs reverse letters
s = "learn python fast"
print(" ".join(s.split()[::-1]), "|", s[::-1])

# 5. any/all over characters
pw = "Secret12"
print(any(c.isdigit() for c in pw), all(c.isalnum() for c in pw))

# 6. Counter: the frequency tool you'll love
from collections import Counter
print(Counter("mississippi").most_common(2))

# 7. textwrap & string module
import textwrap, string
print(textwrap.fill("Python is a wonderful language for learning programming.", width=25))
print(string.ascii_lowercase, string.digits, string.punctuation[:10])

# 8. Multiple separators via regex split (preview of the re module)
import re
print(re.split(r"[,;\s]+", "a,b; c   d"))

# 9. removeprefix / removesuffix (3.9+) -- exact text, unlike strip
print("test_file.py".removesuffix(".py"), "Mr. Smith".removeprefix("Mr. "))

# 10. Unicode surprises
print(len("👍"), "🐍".encode("utf-8"), "ß".upper())
print("straße".casefold() == "STRASSE".casefold())

# 11. Short identical literals may share memory (interning)
a = "hello"; b = "hello"
print(a is b)      # implementation detail! Never rely on `is` for strings.

# 12. Sort by length, then alphabetically, in one key
words = ["pear", "fig", "apple", "kiwi"]
print(sorted(words, key=lambda w: (len(w), w)))

# 13. Dynamic width in f-strings
width = 12
print(f"{'dynamic':>{width}}|")

# 14. Binary strings
print(f"{10:08b}", int("00001010", 2), f"{3.14159:.3}")
````

**Hidden fact:** the compiler *constant-folds* `"a" + "b"` into `"ab"` before your program even runs.

**Hidden fact:** `str` methods are implemented in C, so `"".join(list)` and `s.replace()` beat hand-written loops by 10–50×.
