# 2. ⚡ Syntax shortcuts

```python
# --- Unpacking everywhere ---
a, b, *rest = range(6)
print(a, b, rest)
(x, y), z = (1, 2), 3
first, *_, last = "python"
print(first, last)

# --- Merge / spread ---
d1, d2 = {"a": 1}, {"b": 2}
print({**d1, **d2, "c": 3}, [*range(3), *"ab"], {*"aab", *"bc"})

# --- Chained comparison & conditional expression ---
n = 15
print(10 < n < 20, "big" if n > 10 else "small")

# --- Multiple targets in for/with/comprehension ---
for i, (k, v) in enumerate({"a": 1, "b": 2}.items()):
    print(i, k, v, end=" | ")
print()

# --- f-string power tools ---
name, val, w = "pi", 3.14159265, 10
print(f"{name!r:>8} | {val:{w}.3f} | {val:e} | {1234567:,} | {0.256:.1%} | {255:#010b} | {'x':^7}|")
import datetime
print(f"{datetime.date(2025, 3, 14):%d %b %Y, %A}")
print(f"{val=:.2f}  {name=}")

# --- Walrus := ---
data = [1, 5, 9, 12]
if (n := len(data)) > 3:
    print(f"{n} items")
print([y for x in data if (y := x * 2) > 10])

# --- Underscore uses ---
_, keep = (1, 2)                # throw-away
big = 1_000_000                 # readability
# _private, __mangled, __dunder__, trailing_ (avoid keyword clash: class_)
for _ in range(2): print("hi", end=" ")
print()

# --- Truthy shortcuts ---
items = []
print("empty" if not items else "has items", items or "default", None or 0 or "" or "last")

# --- String tricks ---
print("-" * 20, "=".join("abc"), "a b c".split(), "x".join(["1", "2", "3"]))
print("Python"[::-1], "Python"[::2], "%s is %d" % ("age", 20))

# --- Swap, rotate, transpose ---
x, y = 1, 2; x, y = y, x
print(x, y, list(zip(*[[1, 2, 3], [4, 5, 6]])))

# --- Slicing with slice objects and negative steps ---
L = list(range(10))
print(L[::-1][:3], L[-3:], L[1:-1:3], L[::-3])

# --- Dict/set/list comprehension family and generator expressions ---
print({i: i*i for i in range(4)}, {i % 3 for i in range(9)}, sum(i for i in range(101)))

# --- Ternary chains, and/or shortcuts ---
score = 72
print("A" if score >= 90 else "B" if score >= 70 else "C")

# --- Star args ---
print(*"abc", sep="-")
print(max(3, 9, 4), max([3, 9, 4]), max("abc", key=ord))
```

# 3. 🧩 One-liners worth memorising

```python
from collections import Counter
from functools import reduce
import itertools as it, math, random, string

nums = [4, 8, 15, 16, 23, 42]
text = "the quick brown fox jumps over the lazy dog the end"

print("1. sum of even squares      :", sum(x*x for x in nums if x % 2 == 0))
print("2. flatten a matrix         :", [v for row in [[1,2],[3,4],[5]] for v in row])
print("3. transpose                :", [list(r) for r in zip(*[[1,2,3],[4,5,6]])])
print("4. word frequencies (top 2) :", Counter(text.split()).most_common(2))
print("5. is palindrome            :", (lambda s: s == s[::-1])("racecar"))
print("6. factorial                :", math.prod(range(1, 8)))
print("7. fibonacci (10 terms)     :", (lambda n: reduce(lambda a, _: a + [a[-1] + a[-2]], range(n - 2), [0, 1]))(10))
print("8. primes < 40              :", [n for n in range(2, 40) if all(n % d for d in range(2, int(n**.5) + 1))])
print("9. remove duplicates (order):", list(dict.fromkeys([3, 1, 3, 2, 1])))
print("10. chunk into 2s           :", [nums[i:i+2] for i in range(0, len(nums), 2)])
print("11. max by key              :", max(text.split(), key=len))
print("12. dict from two lists     :", dict(zip("abc", [1, 2, 3])))
print("13. invert a dict           :", {v: k for k, v in {"a": 1, "b": 2}.items()})
print("14. count vowels            :", sum(c in "aeiou" for c in text))
print("15. is anagram              :", sorted("listen") == sorted("silent"))
print("16. gcd / lcm               :", math.gcd(48, 18), math.lcm(4, 6))
print("17. binary of 37            :", bin(37)[2:], format(37, "b"), int("100101", 2))
print("18. all pairs               :", list(it.combinations([1, 2, 3], 2)))
print("19. running total           :", list(it.accumulate(nums)))
print("20. rotate list left by 2   :", nums[2:] + nums[:2])
print("21. digits of a number      :", [int(d) for d in str(90210)], sum(map(int, str(90210))))
print("22. random password         :", len("".join(random.choices(string.ascii_letters + string.digits, k=12))))
print("23. read file lines         : open('f.txt').read().splitlines()")
print("24. find index of max       :", max(range(len(nums)), key=nums.__getitem__))
print("25. second largest          :", sorted(set(nums))[-2])
print("26. sort dict by value      :", dict(sorted({"a": 3, "b": 1, "c": 2}.items(), key=lambda kv: kv[1])))
print("27. group by first letter   :", {k: list(g) for k, g in it.groupby(sorted(["apple", "avocado", "banana", "blueberry"]), key=lambda w: w[0])})
print("28. cartesian product       :", [f"{a}{b}" for a, b in it.product("AB", "12")])
print("29. zip longest w/ fill     :", list(it.zip_longest("abc", "12", fillvalue="_")))
print("30. sum of digits until 1   :", (lambda n: 1 + (n - 1) % 9 if n else 0)(987654))
```

# 4. 🪤 Gotchas — Python's famous traps
Predict each result first. Then read the explanation.

```python
# 1) Mutable default argument
def add(x, lst=[]):
    lst.append(x); return lst
print(add(1), add(2))                     # [1, 2] [1, 2]  -- the SAME list is reused

# 2) List of lists multiplication
grid = [[0] * 2] * 2
grid[0][0] = 9
print(grid)                               # [[9, 0], [9, 0]]  -- both rows are the same object

# 3) Late-binding closures
funcs = [lambda: i for i in range(3)]
print([f() for f in funcs])               # [2, 2, 2]  -- i is looked up when called

# 4) is vs ==
a = 1000; b = 1000
print(a == b, a is b)                     # True, usually False (only small ints are cached)   [in a REPL each line is separate]
# NOTE: inside ONE compiled cell/module, constants may be merged, so `a is b` can show True here.

# 5) Float precision
print(0.1 + 0.2 == 0.3, round(2.675, 2), round(0.5), round(1.5), round(2.5))   # False 2.67 0 2 2

# 6) Modifying a list while iterating
L = [1, 2, 3, 4]
for x in L:
    if x % 2 == 0: L.remove(x)
print(L)                                  # [1, 3]  ... works by luck; try [2, 4, 6] -> [4]
L = [2, 4, 6]
for x in L:
    L.remove(x)
print(L)

# 7) Tuple with a mutable item
t = ([1], 2)
try: t[0] += [3]
except TypeError: pass
print(t)                                  # ([1, 3], 2)  -- the error is raised AFTER the list changed

# 8) Bool is int
print(True + True, sum([True] * 5), {True: "a", 1: "b"})   # 2 5 {True: 'b'}

# 9) Chained comparison surprise
print(1 == 1 in [1], (1 == 1) in [1], 1 == (1 in [1]))     # True True True; but: `False == False in [False]`
print(False == False in [False])         # True ... it means (False == False) and (False in [False])

# 10) Integer division and modulo on negatives
print(-7 // 2, -7 % 2, int(-7 / 2))       # -4 1 -3

# 11) Default `sort` returns None
print([3, 1, 2].sort(), sorted([3, 1, 2]))

# 12) String multiplication of lists of strings & `+=` on lists
x = [1]; y = x; y += [2]; print(x)        # [1, 2]  (in-place extend)
x = (1,); y = x; y += (2,); print(x, y)   # (1,) (1, 2)  (tuples are immutable -> new object)

# 13) `and`/`or` return operands
print(0 or [] or "x", 1 and 2 and 3, "" and "y")

# 14) Class attribute vs instance attribute
class C: items = []
c1, c2 = C(), C(); c1.items.append(1)
print(c2.items)                           # [1]  -- shared

# 15) except (A, B) needs parentheses, and `except A, B:` was Python 2
# 16) round() uses banker's rounding: round(0.5) == 0, round(1.5) == 2, round(2.5) == 2
# 17) `x = y = []` makes ONE list. `a, b = [], []` makes two.
# 18) Strings: "a" "b" adjacent literal concatenation inside a list forgot-comma bug:
print(["a", "b" "c", "d"])                # ['a', 'bc', 'd']  <- missing comma joins strings!
# 19) A function without return gives None, and `print(print("x"))` prints None
# 20) `datetime.utcnow()` is naive and deprecated -> use datetime.now(timezone.utc)
```

# 5. 🏎️ Performance secrets

```python
import timeit

def bench(label, stmt, setup="pass", n=2000):
    t = timeit.timeit(stmt, setup, number=n) / n * 1e6
    print(f"{label:38} {t:9.2f} µs")

setup = "data = list(range(1000)); s = set(data); d = {i: i for i in data}"
bench("membership: x in list", "999 in data", setup)
bench("membership: x in set", "999 in s", setup)
print()
bench("build list: for + append", "r=[]\nfor i in range(1000): r.append(i*2)")
bench("build list: comprehension", "[i*2 for i in range(1000)]")
bench("build list: map(lambda)", "list(map(lambda i: i*2, range(1000)))")
print()
bench("string concat in loop (+=)", "s=''\nfor i in range(1000): s += str(i)")
bench("string build with join", "''.join(str(i) for i in range(1000))")
print()
bench("sum with loop", "t=0\nfor i in range(1000): t+=i")
bench("sum() builtin", "sum(range(1000))")
print()
bench("dict.get vs try/except (hit)", "d.get(5)", setup)
bench("copy list: data[:]", "data[:]", setup)
bench("copy list: list(data)", "list(data)", setup)
bench("{} literal", "{}")
bench("dict() call", "dict()")
```

### The speed rules of thumb
1. **Use the right data structure** (`set`/`dict` lookups ≫ list scans).
2. **Use built-ins** (`sum`, `min`, `sorted`, `str.join`) – they run in C.
3. **Comprehensions/generators** beat manual append loops.
4. **Avoid work in loops**: hoist constants, cache attribute lookups, precompute.
5. **Cache results**: `functools.cache` / `lru_cache`.
6. **Vectorise** with NumPy/pandas for numeric work.
7. **Profile first** (`cProfile`, `timeit`), optimise the 1% that matters.
8. For real heavy lifting: Numba, Cython, PyPy, or move the hot loop to Rust/C.

```python
from functools import lru_cache
import time

def collatz_len(n):
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps

@lru_cache(maxsize=None)
def collatz_cached(n):
    return 0 if n == 1 else 1 + collatz_cached(n // 2 if n % 2 == 0 else 3 * n + 1)

t = time.perf_counter(); best = max(range(1, 30000), key=collatz_len); t1 = time.perf_counter() - t
t = time.perf_counter(); best2 = max(range(1, 30000), key=collatz_cached); t2 = time.perf_counter() - t
print(best, best2, f"plain {t1:.2f}s  cached {t2:.2f}s")
```

```python
# A tiny homemade debugging helper (icecream-lite)
import inspect
def ic(x):
    frame = inspect.currentframe().f_back
    ctx = inspect.getframeinfo(frame).code_context      # None if the source isn't available
    label = ctx[0].strip().split("ic(", 1)[1].rsplit(")", 1)[0] if ctx else "value"
    print(f"ic| {label}: {x!r}")
    return x

total = ic(sum(range(10)))
words = ic("hello".upper())
```

```python
# Pretty printing nested data and a quick timer context manager
from pprint import pprint
from contextlib import contextmanager
import time

pprint({"users": [{"name": "Asha", "roles": ["admin", "dev"]}, {"name": "Ravi", "roles": []}], "count": 2}, width=50)

@contextmanager
def stopwatch(label="block"):
    t = time.perf_counter()
    yield
    print(f"[{label}] {time.perf_counter() - t:.4f}s")

with stopwatch("sum"):
    sum(range(1_000_000))
```

# 7. 🧠 How to learn (and debug) faster

### The learning loop
1. **Read** a concept → 2. **Type it yourself** (never copy-paste) → 3. **Break it** on purpose → 4. **Explain it** aloud (rubber-duck) → 5. **Build something** tiny with it.

### 30-second debugging checklist
1. Read the **last line** of the traceback first.
2. Print the **type and value**: `print(type(x), repr(x))`.
3. Reproduce with the **smallest** example.
4. Check the assumptions: is the list empty? Is it a `str` instead of `int`? Off-by-one?
5. Use `breakpoint()` and step through.
6. Google the exact error message (remove the file paths). Read the docs: `python -m pydoc modulename`.
7. Take a break – bugs are often obvious after a walk.

### Studying for exams
- **Active recall**: cover the answer, write the code from memory.
- **Spaced repetition**: revisit topics after 1, 3, 7, 14 days.
- **Trace by hand**: write variable values line by line for output questions.
- **Time yourself** on the mock papers in `exam_prep/`.
- Know the *why*: examiners love "differentiate between…" questions.
