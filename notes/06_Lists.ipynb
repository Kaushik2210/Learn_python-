## 1. What is a list?
An **ordered, mutable, heterogeneous** collection written with square brackets. Duplicates allowed. Under the hood it is a *dynamic array of references*.

````python
nums = [10, 20, 30, 40]
mixed = [1, "two", 3.0, [4, 5], True, None]
empty = []
from_range = list(range(5))
from_string = list("hey")

print(nums, mixed, empty, from_range, from_string, sep="\n")
print(len(mixed), type(nums))
````

## 2. Indexing & slicing (exactly like strings)

````python
L = [10, 20, 30, 40, 50, 60]
print(L[0], L[-1], L[2])
print(L[1:4], L[:3], L[3:], L[::2], L[::-1])
print(L[-3:], L[:-1])

nested = [1, [2, 3, [4, 5]]]
print(nested[1][2][0])        # 4  -> chain indexes to go deeper
````

## 3. Lists are mutable — changing them

````python
L = [1, 2, 3, 4, 5]
L[0] = 100                 # change one item
L[1:3] = [20, 30, 35]      # replace a slice (sizes needn't match!)
print(L)
L[1:3] = []                # delete a slice
print(L)
L[1:1] = ["a", "b"]        # insert at position 1 without removing anything
print(L)
````

## 4. List methods (learn ALL of these)

| Method | What it does | Returns | Time |
|---|---|---|---|
| `append(x)` | add to end | None | O(1) |
| `extend(iterable)` | add all items of iterable | None | O(k) |
| `insert(i, x)` | insert before index i | None | O(n) |
| `remove(x)` | delete first occurrence of value | None | O(n) |
| `pop([i])` | remove & return item (default last) | item | O(1) last / O(n) elsewhere |
| `clear()` | remove everything | None | O(n) |
| `index(x)` | position of first x (ValueError if absent) | int | O(n) |
| `count(x)` | occurrences of x | int | O(n) |
| `sort()` | sort in place | None | O(n log n) |
| `reverse()` | reverse in place | None | O(n) |
| `copy()` | shallow copy | list | O(n) |

> ⚠️ Most mutating methods return **`None`**. `x = mylist.sort()` makes `x` equal `None` — a very common bug!

````python
L = [3, 1, 2]
L.append(4);          print("append   ", L)
L.extend([5, 6]);     print("extend   ", L)
L.insert(0, 99);      print("insert   ", L)
L.remove(99);         print("remove   ", L)
print("pop()    ", L.pop(), L)
print("pop(0)   ", L.pop(0), L)
print("index(2) ", L.index(2), " count(2):", L.count(2))
L.reverse();          print("reverse  ", L)
L.sort();             print("sort     ", L)
L.clear();            print("clear    ", L)

# append vs extend
a = [1, 2]; a.append([3, 4]);  print(a)      # [1, 2, [3, 4]]
b = [1, 2]; b.extend([3, 4]);  print(b)      # [1, 2, 3, 4]
c = [1, 2]; c += [3, 4];       print(c)      # += behaves like extend
print([1, 2] + [3], [0] * 4)
````

### Deleting: `del`, `remove`, `pop`
- `del L[i]` / `del L[a:b]` – by position (no return)
- `L.remove(x)` – by value
- `L.pop(i)` – by position AND returns the value

````python
L = list("abcdef")
del L[0]; print(L)
del L[1:3]; print(L)
del L[::2]; print(L)
````

## 5. Membership, min/max/sum, and friends

````python
L = [4, 8, 15, 16, 23, 42]
print(15 in L, 99 not in L)
print(len(L), min(L), max(L), sum(L), sum(L) / len(L))
print(sorted(L, reverse=True))
print(any(x > 40 for x in L), all(x > 3 for x in L))
print(list(reversed(L)))
````

## 6. Sorting

- `L.sort()` – sorts **in place**, returns `None`
- `sorted(L)` – returns a **new** sorted list, works on any iterable
- `key=` – a function applied to each element to decide its sort position
- `reverse=True` – descending
- Sort is **stable**: equal items keep their original order (so you can sort by several keys in passes)

````python
words = ["banana", "Apple", "cherry", "date"]
print(sorted(words))                          # uppercase first (ASCII)
print(sorted(words, key=str.lower))           # case-insensitive
print(sorted(words, key=len))                 # by length
print(sorted(words, key=len, reverse=True))

students = [("Asha", 88), ("Ravi", 92), ("Meena", 88), ("Tom", 75)]
print(sorted(students, key=lambda s: s[1], reverse=True))
print(sorted(students, key=lambda s: (-s[1], s[0])))   # score desc, then name asc

from operator import itemgetter
print(sorted(students, key=itemgetter(1)))
````

## 7. Copying lists — the most important gotcha

````python
a = [1, 2, 3]
b = a              # NOT a copy: two names, one list
b.append(99)
print(a, b, a is b)

c = a.copy()       # shallow copy (also: a[:], list(a))
c.append(0)
print(a, c, a is c)
````

### Shallow vs deep copy
A shallow copy copies the outer list, but the **inner objects are shared**.

````python
import copy
matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
deep = copy.deepcopy(matrix)

matrix[0][0] = 999
print("original:", matrix)
print("shallow :", shallow)   # sees the change!
print("deep    :", deep)      # independent

# The classic bug: multiplying nested lists
bad = [[0] * 3] * 3       # three references to the SAME row
bad[0][0] = 1
print(bad)                # [[1,0,0],[1,0,0],[1,0,0]]  <- surprise!

good = [[0] * 3 for _ in range(3)]
good[0][0] = 1
print(good)
````

## 8. List comprehensions

`[expression for item in iterable if condition]` — a compact way to build lists.

````python
squares = [x ** 2 for x in range(1, 8)]
evens   = [x for x in range(20) if x % 2 == 0]
upper   = [w.upper() for w in ["a", "b"]]
labels  = ["even" if x % 2 == 0 else "odd" for x in range(5)]  # if-else goes BEFORE for
pairs   = [(x, y) for x in range(3) for y in range(3) if x != y]
flat    = [v for row in [[1, 2], [3, 4]] for v in row]
lengths = {w: len(w) for w in ["hi", "python"]}      # dict comprehension
uniq    = {c for c in "hello"}                       # set comprehension

print(squares, evens, upper, labels, pairs, flat, lengths, sorted(uniq), sep="\n")
````

> Read comprehensions as: **"give me *expression* for each *item* in *iterable* (that passes *condition*)"**.

Filtering forms: `[x for x in data if cond]` (filter). Transforming: `[f(x) for x in data]` (map). Never put more than two `for`s or complicated logic in one — use a normal loop for readability.

## 9. 2-D lists (matrices)

````python
m = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]

print(m[1][2])                                  # row 1, col 2 -> 6
for row in m:
    print(*row)

print("row sums :", [sum(r) for r in m])
print("col sums :", [sum(col) for col in zip(*m)])          # zip(*m) transposes!
print("transpose:", [list(c) for c in zip(*m)])
print("diagonal :", [m[i][i] for i in range(3)])
print("flatten  :", [x for r in m for x in r])

# Matrix multiplication (manual)
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
C = [[sum(a * b for a, b in zip(row, col)) for col in zip(*B)] for row in A]
print(C)
````

## 10. Classic list programs

````python
data = [5, 3, 8, 3, 9, 1, 5, 8]

# second largest
print(sorted(set(data))[-2])

# remove duplicates, preserving order
seen, unique = set(), []
for x in data:
    if x not in seen:
        seen.add(x); unique.append(x)
print(unique, list(dict.fromkeys(data)))

# frequency
from collections import Counter
print(Counter(data))

# rotate left by k
def rotate(lst, k):
    k %= len(lst)
    return lst[k:] + lst[:k]
print(rotate([1, 2, 3, 4, 5], 2))

# linear search
def linear_search(lst, target):
    for i, v in enumerate(lst):
        if v == target:
            return i
    return -1
print(linear_search(data, 9), linear_search(data, 100))

# bubble sort
def bubble_sort(a):
    a = a[:]
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        if not swapped:
            break
    return a
print(bubble_sort(data))

# merge two sorted lists
def merge(a, b):
    i = j = 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i]); i += 1
        else:
            out.append(b[j]); j += 1
    return out + a[i:] + b[j:]
print(merge([1, 4, 9], [2, 3, 10]))

# common elements / union / difference (with sets)
a, b = [1, 2, 3, 4], [3, 4, 5]
print(sorted(set(a) & set(b)), sorted(set(a) | set(b)), sorted(set(a) - set(b)))

# separate even and odd
evens = [x for x in data if x % 2 == 0]
odds = [x for x in data if x % 2]
print(evens, odds)

# two-sum with a dict: O(n)
def two_sum(nums, target):
    seen = {}
    for i, n in enumerate(nums):
        if target - n in seen:
            return seen[target - n], i
        seen[n] = i
print(two_sum([2, 7, 11, 15], 9))
````

## 11. Lists as stacks and queues
- **Stack** (LIFO): `append()` + `pop()` – both O(1) ✅
- **Queue** (FIFO): `append()` + `pop(0)` – `pop(0)` is O(n) ❌ → use `collections.deque`

````python
stack = []
stack.append(1); stack.append(2); stack.append(3)
print(stack.pop(), stack)

from collections import deque
q = deque([1, 2, 3])
q.append(4)
print(q.popleft(), q)
q.rotate(1); print(q)
````

# 🕵️ Hidden Facts & Pro Tricks

````python
# 1. Unpacking with * : split head and tail, or merge lists
first, *rest = [1, 2, 3, 4]
print(first, rest)
merged = [*[1, 2], *(3, 4), *"56"]
print(merged)

# 2. Swap two elements
a = [1, 2, 3]
a[0], a[-1] = a[-1], a[0]
print(a)

# 3. Slice assignment with step (must match the size)
a = list(range(10))
a[::2] = ["e"] * 5
print(a)

# 4. Trick: transpose, flatten, rotate matrix 90 degrees clockwise
m = [[1, 2, 3], [4, 5, 6]]
print([list(r) for r in zip(*m[::-1])])

# 5. Chunk a list into groups of n
lst = list(range(1, 11))
print([lst[i:i + 3] for i in range(0, len(lst), 3)])

# 6. min/max with key, and getting the INDEX of the max
vals = [4, 9, 2, 9]
print(max(range(len(vals)), key=vals.__getitem__), max(["aa", "b", "cccc"], key=len))

# 7. Lists over-allocate: append is amortised O(1)
import sys
sizes = set()
L = []
for i in range(30):
    L.append(i)
    sizes.add(sys.getsizeof(L))
print(sorted(sizes))       # memory grows in jumps, not on every append

# 8. `sum` of lists flattens (but is slow, O(n^2)); prefer itertools.chain
print(sum([[1], [2, 3]], []))

# 9. Lists compare lexicographically; you can sort lists of lists
print(sorted([[2, 1], [1, 5], [1, 2]]))

# 10. Default mutable argument trap (memory hook for functions module)
def bad(x, acc=[]):
    acc.append(x)
    return acc
print(bad(1), bad(2))      # [1, 2]  <- shared between calls!

# 11. Removing while iterating: use comprehension instead
L = [1, 2, 3, 4, 5, 6]
L = [x for x in L if x % 2]
print(L)

# 12. enumerate + comprehension: indices of matches
print([i for i, v in enumerate("banana") if v == "a"])

# 13. List multiplication with immutable items is safe
print([None] * 3, ["-"] * 5)

# 14. `list.sort` is Timsort: adaptive - nearly-sorted data is sorted in ~O(n)
import timeit
print(timeit.timeit("sorted(a)", "a=list(range(100000))", number=20) <
      timeit.timeit("sorted(a)", "import random; a=random.sample(range(100000),100000)", number=20))

# 15. Use bisect for fast sorted-list search/insert
import bisect
s = [10, 20, 30, 40]
bisect.insort(s, 25)
print(s, bisect.bisect_left(s, 30))

# 16. array/deque/numpy when list isn't the best tool:
#     deque for queues, array('i') for compact numbers, numpy for maths.
````

**Hidden fact:** a Python list stores *pointers*, not the values. That's why a list of 1M ints takes ~8 MB for pointers + ~28 MB for the int objects, while a NumPy array of 1M int64 takes exactly 8 MB.

**Hidden fact:** `[] == []` is `True` but `[] is []` is `False` — each `[]` builds a new object. And **`x = y = []` makes ONE list with two names**.
