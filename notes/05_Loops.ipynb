## 1. Why loops?
Repeating code by copy-paste is bad. A **loop** repeats a block while a condition holds, or once per item of a collection.

Python has two loops: **`while`** (repeat while a condition is true) and **`for`** (repeat for each item in an iterable).

## 2. The `while` loop

```
while condition:
    body
```
The condition is checked **before each iteration**. Three ingredients for a safe loop: **initialise → test → update**.

````python
count = 1                  # 1) initialise
while count <= 5:          # 2) test
    print("Count is", count)
    count += 1             # 3) update (forget this => infinite loop!)
print("Done")
````

````python
# Sum of digits using while
n = 4729
total = 0
while n > 0:
    total += n % 10        # last digit
    n //= 10               # drop last digit
print("Digit sum:", total)
````

````python
# Sentinel-controlled loop: repeat until the user says stop
# (input is faked to keep returning '5' in this demo, so we simulate with a list)
data = [3, 8, 2, -1, 99]   # -1 is the sentinel (stop signal)
i = 0
running_total = 0
while data[i] != -1:
    running_total += data[i]
    i += 1
print("Total before sentinel:", running_total)
````

### Infinite loops
`while True:` loops forever unless you `break`. That's a legitimate pattern for menus and input validation.

````python
attempts = 0
while True:
    attempts += 1
    if attempts * attempts > 50:
        break
print("stopped after", attempts, "attempts")
````

## 3. The `for` loop

```
for item in iterable:
    body
```
An **iterable** is anything you can loop over: strings, lists, tuples, sets, dicts, ranges, files, generators…

````python
for ch in "Hey":
    print(ch)

for fruit in ["apple", "banana", "cherry"]:
    print(fruit.upper())

for x in (1, 2, 3):
    print(x * x, end=" ")
print()

for key in {"a": 1, "b": 2}:      # iterating a dict gives keys
    print(key, end=" ")
````

### 3.1 `range()` — the number generator

| Call | Produces |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 |

`range(start, stop, step)` — **stop is excluded**. It's lazy (doesn't build the whole list), so `range(10**9)` costs no memory.

````python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(0, 10, 2)))
print(list(range(10, 0, -2)))
print(list(range(5, 0)))          # [] – empty because step is +1
r = range(1, 100, 3)
print(len(r), r[0], r[-1], 40 in r)

for i in range(3):
    print("Iteration", i)
````

### 3.2 Looping with the index: `enumerate`

````python
colors = ["red", "green", "blue"]

# ❌ un-Pythonic
for i in range(len(colors)):
    print(i, colors[i])

# ✅ Pythonic
for i, color in enumerate(colors):
    print(i, color)

for i, color in enumerate(colors, start=1):    # custom start
    print(f"{i}. {color}")
````

### 3.3 Looping over two sequences: `zip`

````python
names = ["Asha", "Ravi", "Meena"]
marks = [88, 92, 79]

for n, m in zip(names, marks):
    print(f"{n}: {m}")

print(list(zip("abc", [1, 2, 3, 4])))     # stops at the shortest
print(dict(zip(names, marks)))

# Unzipping
pairs = [("a", 1), ("b", 2)]
letters, numbers = zip(*pairs)
print(letters, numbers)
````

### 3.4 `reversed`, `sorted`

````python
for x in reversed([1, 2, 3]):
    print(x, end=" ")
print()
for x in sorted([3, 1, 2], reverse=True):
    print(x, end=" ")
print()
for ch in reversed("abc"):
    print(ch, end="")
````

## 4. Loop control statements

| Statement | Effect |
|---|---|
| `break` | exit the **innermost** loop immediately |
| `continue` | skip the rest of this iteration; go to the next |
| `pass` | do nothing (placeholder) |

````python
# break: stop at first negative
for n in [4, 9, -2, 7]:
    if n < 0:
        print("negative found, stopping")
        break
    print("ok", n)

# continue: skip odd numbers
for n in range(1, 8):
    if n % 2 == 1:
        continue
    print("even", n)
````

### The loop `else` clause (unique to Python!) ⭐
The `else` block runs **only if the loop finished normally (no `break`)**. Perfect for "search & not found".

````python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break                 # found a divisor
    else:
        return True               # no break => no divisor => prime
    return False

print([n for n in range(2, 40) if is_prime(n)])

# Same on a while loop
i = 0
while i < 3:
    i += 1
else:
    print("while finished without break, i =", i)
````

## 5. Nested loops
A loop inside a loop. Inner loop runs *completely* for each iteration of the outer loop. Total iterations = outer × inner.

````python
# Multiplication table 1..5
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i*j:3}", end="")
    print()                    # newline after each row
````

### 🎨 Pattern programs (a favourite exam question)
Rule of thumb: **outer loop = rows, inner loop = columns**. Work out how many spaces/stars per row.

````python
n = 5

print("Right triangle:")
for i in range(1, n + 1):
    print("*" * i)

print("\nInverted triangle:")
for i in range(n, 0, -1):
    print("*" * i)

print("\nPyramid:")
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))

print("\nNumber triangle:")
for i in range(1, n + 1):
    print(" ".join(str(j) for j in range(1, i + 1)))

print("\nFloyd's triangle:")
num = 1
for i in range(1, 5):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()

print("\nDiamond:")
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))

print("\nHollow square:")
for i in range(n):
    for j in range(n):
        print("*" if i in (0, n - 1) or j in (0, n - 1) else " ", end=" ")
    print()

print("\nAlphabet triangle:")
for i in range(1, 5):
    print("".join(chr(64 + j) for j in range(1, i + 1)))
````

## 6. Classic loop programs

````python
# 1. Factorial
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
print(factorial(5), factorial(0), factorial(10))

# 2. Fibonacci (first n terms)
def fib(n):
    a, b = 0, 1
    out = []
    for _ in range(n):
        out.append(a)
        a, b = b, a + b
    return out
print(fib(10))

# 3. Reverse a number
def reverse_number(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n //= 10
    return rev
print(reverse_number(12345))

# 4. Palindrome number
def is_pal_num(n):
    return n == reverse_number(n)
print(is_pal_num(12321), is_pal_num(123))

# 5. Armstrong number (sum of digits^len == n)
def is_armstrong(n):
    digits = str(n)
    return n == sum(int(d) ** len(digits) for d in digits)
print([n for n in range(100, 1000) if is_armstrong(n)])

# 6. GCD (Euclid)
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
print(gcd(48, 18))

# 7. Perfect number (sum of proper divisors == n)
def is_perfect(n):
    return n > 1 and sum(d for d in range(1, n) if n % d == 0) == n
print([n for n in range(1, 500) if is_perfect(n)])

# 8. Collatz sequence
def collatz(n):
    steps = [n]
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps.append(n)
    return steps
print(collatz(6))

# 9. Sum of series 1 + 1/2 + 1/3 + ... + 1/n
print(round(sum(1 / i for i in range(1, 11)), 4))

# 10. Count digits / binary conversion by hand
def to_binary(n):
    if n == 0:
        return "0"
    bits = ""
    while n > 0:
        bits = str(n % 2) + bits
        n //= 2
    return bits
print(to_binary(37), bin(37)[2:])

# 11. FizzBuzz (the famous interview question)
for i in range(1, 16):
    print("FizzBuzz" if i % 15 == 0 else "Fizz" if i % 3 == 0 else "Buzz" if i % 5 == 0 else i, end=" ")
print()

# 12. Number guessing logic (binary search idea)
lo, hi, target, tries = 1, 100, 73, 0
while lo <= hi:
    mid = (lo + hi) // 2
    tries += 1
    if mid == target:
        break
    lo, hi = (mid + 1, hi) if mid < target else (lo, mid - 1)
print("found", target, "in", tries, "tries")
````

## 7. Loop tricks & idioms

````python
# Accumulator patterns
total, biggest, evens = 0, float("-inf"), []
for n in [5, 12, 7, 20, 3]:
    total += n
    biggest = max(biggest, n)
    if n % 2 == 0:
        evens.append(n)
print(total, biggest, evens)

# Looping over dict items
scores = {"Asha": 90, "Ravi": 72}
for name, sc in scores.items():
    print(f"{name} -> {sc}")

# Loop variable persists after the loop
for k in range(3):
    pass
print("k after loop:", k)

# Don't modify a list while iterating over it
nums = [1, 2, 3, 4, 5, 6]
for n in nums[:]:               # iterate over a COPY
    if n % 2 == 0:
        nums.remove(n)
print(nums)

# Underscore for throw-away variable
for _ in range(2):
    print("hi")

# Flattening with nested loop
matrix = [[1, 2], [3, 4], [5, 6]]
flat = []
for row in matrix:
    for v in row:
        flat.append(v)
print(flat)
````

## 8. Common mistakes

| Mistake | Symptom | Fix |
|---|---|---|
| Not updating the loop variable in `while` | infinite loop | make sure the condition can become False |
| Off-by-one with `range` | missing last item | remember stop is **excluded** |
| Using `range(len(x))` just to index | clunky | use `for item in x` or `enumerate` |
| Changing a list while looping over it | skipped items | iterate over a copy |
| `break` in nested loops | only inner loop exits | use a flag or wrap in a function and `return` |

## ✅ Summary
- `while` = repeat while a condition holds; `for` = repeat for each item.
- `range(start, stop, step)`; stop excluded; lazy.
- `break` exits, `continue` skips, `else` runs when no `break` happened.
- Use `enumerate` for index+value and `zip` for parallel iteration.
- Patterns: outer loop = rows, inner loop = columns.

# 🕵️ Hidden Facts & Pro Tricks (Module 05)

````python
# 1. range is a real sequence: O(1) membership, indexing, len
r = range(0, 10**12, 7)
print(len(r), 14 in r, 15 in r, r[3], r[-1])

# 2. iter() with a sentinel: loop until a value appears
import random
random.seed(1)
rolls = iter(lambda: random.randint(1, 6), 6)     # stops when it gets a 6
print(list(rolls))

# 3. itertools: the loop power tools
import itertools as it
print(list(it.combinations("ABC", 2)))
print(list(it.permutations([1, 2, 3], 2)))
print(list(it.product("ab", repeat=2)))
print(list(it.islice(it.count(10, 5), 4)))
print(list(it.accumulate([1, 2, 3, 4])))               # running totals
print([(k, len(list(g))) for k, g in it.groupby("aaabbc")])
print(list(it.chain([1, 2], (3, 4), "56")))
print(list(it.zip_longest("ab", [1, 2, 3], fillvalue="-")))
print(list(it.pairwise([1, 4, 9, 16])))                # 3.10+
print(list(it.batched(range(7), 3)))                   # 3.12+

# 4. Compare neighbours with zip(seq, seq[1:])
seq = [10, 20, 35]
for cur, nxt in zip(seq, seq[1:]):
    print(nxt - cur, end=" ")
print()

# 5. Dict/set comprehensions replace loops
print({c: ord(c) for c in "abc"}, {x % 3 for x in range(10)})

# 6. Nested loops in one comprehension
print([(i, j) for i in range(3) for j in range(i)])

# 7. sum/min/max/any/all accept generators (memory-light)
print(sum(i * i for i in range(10**6)))

# 8. Ranking with enumerate + sorted
for rank, (name, score) in enumerate(sorted({"A": 3, "B": 9, "C": 5}.items(), key=lambda kv: -kv[1]), 1):
    print(rank, name, score)

# 9. Break out of nested loops cleanly using a function
def find_pair(grid, target):
    for i, row in enumerate(grid):
        for j, v in enumerate(row):
            if v == target:
                return i, j
    return None
print(find_pair([[1, 2], [3, 4]], 3))
````

**Hidden fact — the `for` loop is secretly `while True` + `next()`:**

```python
it = iter(iterable)
while True:
    try:
        item = next(it)
    except StopIteration:
        break
    ...body...
```

````python
it = iter([1, 2, 3])
print(next(it), next(it), next(it), next(it, "exhausted"))
````

**Hidden fact — speed:** a comprehension is roughly 30% faster than the equivalent `for`+`append` loop, and `sum(generator)` uses almost no memory. But a plain `for` loop is fine for clarity — optimise only when it matters.
