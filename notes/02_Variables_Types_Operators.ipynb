## 1. Variables

A **variable** is a *name* that refers to a value stored in memory. You create one by assigning with `=`. No declaration, no type needed.

```python
name = "Asha"
age = 19
height = 1.62
is_student = True

print(name, age, height, is_student)
```

### 🧠 The right mental model: names are labels, not boxes
In C, a variable is a *box* holding a value. In Python, a variable is a **label (reference) stuck onto an object**. Several labels can point to the same object.

```python
a = [1, 2, 3]
b = a            # b is another label on the SAME list
b.append(4)
print(a)         # a sees the change!
print(a is b)    # True: same object
print(id(a), id(b))
```

### Dynamic typing
The *value* has a type; the *name* can be re-pointed to anything.

```python
x = 10
print(x, type(x))
x = "ten"
print(x, type(x))
x = [1, 0]
print(x, type(x))
```

### Multiple assignment tricks

```python
a = b = c = 0                 # all three -> 0
x, y, z = 1, 2, 3             # unpacking
print(a, b, c, x, y, z)

x, y = y, x                   # swap without a temp variable!
print(x, y)

first, *middle, last = [10, 20, 30, 40, 50]   # star-unpacking
print(first, middle, last)
```

### Naming rules & conventions (recap)
- letters, digits, `_`; can't start with a digit; no keywords; case-sensitive
- `snake_case` for variables/functions, `UPPER_CASE` for constants (Python has no true constants — it's a convention)
- Use meaningful names: `total_price` ≫ `tp` ≫ `x`

```python
MAX_USERS = 100      # constant by convention
_private = "hint: internal use"
__dunder__ = "special names reserved by Python"

# Deleting a name
temp = 5
del temp
try:
    print(temp)
except NameError as e:
    print("NameError:", e)
```

### 2.1 `int` — whole numbers
Python ints have **unlimited precision** (no overflow!).

```python
big = 2 ** 200
print(big)
print(type(big))

# Different bases
print(0b1010)    # binary  -> 10
print(0o17)      # octal   -> 15
print(0xFF)      # hex     -> 255
print(1_000_000) # underscores for readability

# Converting to other bases
n = 45
print(bin(n), oct(n), hex(n))
print(int("101101", 2), int("2D", 16))
```

### 2.2 `float` — decimal numbers
Stored as 64-bit IEEE-754 doubles. **Not exact!**

```python
print(0.1 + 0.2)                 # 0.30000000000000004  (!)
print(0.1 + 0.2 == 0.3)          # False
print(abs((0.1 + 0.2) - 0.3) < 1e-9)   # correct way: compare with tolerance

import math
print(math.isclose(0.1 + 0.2, 0.3))    # nicer

print(1e3, 2.5e-3)               # scientific notation
print(float("inf"), float("-inf"), float("nan"))
```

> **Why?** 0.1 has no exact binary representation, just like 1/3 has no exact decimal one. For money use the `decimal` module:

```python
from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))       # 0.3 exactly
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))
```

### 2.3 `complex` — a + bj
Python has built-in complex numbers (`j`, not `i`).

```python
z = 3 + 4j
print(z, type(z))
print(z.real, z.imag)
print(abs(z))            # magnitude = sqrt(3²+4²) = 5.0
print(z.conjugate())
print(z * (1 - 2j))
```

### 2.4 `bool` — True / False
`bool` is a *subclass of int*: `True == 1`, `False == 0`.

```python
print(True + True)          # 2
print(isinstance(True, int))  # True
print(type(5 > 3))

# Truthiness: what counts as False?
falsy = [0, 0.0, 0j, "", [], (), {}, set(), None, False]
for v in falsy:
    print(f"{v!r:8} -> {bool(v)}")
print(bool("False"), bool(" "), bool([0]), bool(-1))   # all True!
```

**Falsy values** (memorise): `0`, `0.0`, `0j`, `""`, `[]`, `()`, `{}`, `set()`, `range(0)`, `None`, `False`. Everything else is **truthy**.

## 3. Inspecting types: `type()`, `isinstance()`, `id()`

- `type(x)` → exact type
- `isinstance(x, T)` → is `x` an instance of `T` (or a subclass)? **Preferred** for checks.
- `id(x)` → identity (memory address in CPython)

```python
x = 42
print(type(x))
print(type(x) is int)
print(isinstance(x, int))
print(isinstance(True, int))            # True because bool subclasses int
print(type(True) is int)                # False
print(isinstance(x, (int, float)))      # tuple of types = "any of these"
```

## 4. Type conversion (casting)

### Implicit (automatic) conversion
Python widens types when mixing: `int` → `float` → `complex`.

```python
print(5 + 2.0, type(5 + 2.0))        # int + float = float
print(True + 2.5)                    # bool -> float
print((1 + 2j) + 3, type((1+2j)+3))
```

### Explicit conversion (you call the constructor)

| Function | Effect |
|---|---|
| `int(x)` | to integer (truncates floats toward 0) |
| `float(x)` | to float |
| `str(x)` | to string |
| `bool(x)` | to boolean |
| `complex(a, b)` | to complex |
| `list()`, `tuple()`, `set()`, `dict()` | to collections |
| `chr(n)` / `ord(c)` | int ↔ character |

```python
print(int(3.99), int(-3.99))          # 3 -3  (truncation, NOT rounding)
print(int("42"), float("3.5"), float(7))
print(str(99) + "!", str(3.0))
print(bool(0), bool(7))
print(int(True), float(False))
print(chr(65), ord("a"))
print(list("abc"), tuple([1, 2]), set([1, 1, 2]))

# Invalid conversions raise ValueError
for bad in ["3.5", "abc", ""]:
    try:
        print(int(bad))
    except ValueError as e:
        print("ValueError:", e)

print(int(float("3.5")))     # two-step conversion works
```

### Rounding functions

```python
import math
x = 2.5
print(round(x), round(3.5), round(-2.5))   # banker's rounding: .5 goes to nearest EVEN
print(round(3.14159, 2))
print(math.floor(2.7), math.ceil(2.1), math.trunc(-2.7))
print(math.floor(-2.7), math.ceil(-2.1))
```

## 5. Operators

### 5.1 Arithmetic operators

| Op | Name | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | true division (always `float`) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | modulus (remainder) | `7 % 2` | `1` |
| `**` | exponent | `7 ** 2` | `49` |

```python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b)       # 3.4
print(a // b)      # 3
print(a % b)       # 2
print(a ** b)      # 1419857
print(4 / 2)       # 2.0  <- still a float!
```

### ⚠️ Floor division and modulus with negative numbers
`//` rounds toward **negative infinity**, and `%` takes the sign of the **divisor**. The identity `a == (a // b) * b + (a % b)` always holds.

```python
print(-7 // 2)      # -4  (NOT -3)
print(-7 % 2)       #  1  (NOT -1)
print(7 // -2)      # -4
print(7 % -2)       # -1
print(divmod(-7, 2))   # (-4, 1)  quotient and remainder together

a, b = -7, 2
print(a == (a // b) * b + (a % b))
```

### 5.2 Comparison (relational) operators — return `bool`
`==`, `!=`, `>`, `<`, `>=`, `<=`

They can be **chained**: `1 < x < 10` is legal and reads like maths.

```python
x = 5
print(x == 5, x != 5, x > 3, x <= 4)
print(1 < x < 10)          # chained: (1 < x) and (x < 10)
print(1 < x > 3)           # also legal
print("apple" < "banana")  # strings compare lexicographically (by Unicode code points)
print("Z" < "a")           # True: 'Z'=90, 'a'=97
print([1, 2, 3] < [1, 2, 4])  # lists compare element by element
```

### 5.3 Logical operators: `and`, `or`, `not`

- `A and B` → if A is falsy return **A**, else return **B**
- `A or B`  → if A is truthy return **A**, else return **B**
- `not A`   → `True`/`False`

They **short-circuit**: the right side is only evaluated if needed. They return one of the *operands*, not necessarily a bool!

```python
print(True and False, True or False, not True)
print(0 and 5)           # 0   (first is falsy -> returned)
print(3 and 5)           # 5   (first truthy -> returns second)
print(0 or "default")    # 'default'
print("hi" or "default") # 'hi'
print(None or [] or 0)   # 0 (last operand if all falsy)

# Short-circuit demo: the division is never evaluated
x = 0
print(x != 0 and 10 / x > 1)

# Practical idiom: default values
username = ""
display = username or "Guest"
print(display)
```

### 5.4 Assignment operators
`=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, `&=`, `|=`, `^=`, `>>=`, `<<=`, and the walrus `:=` (Python 3.8+).

```python
n = 10
n += 5;  print(n)    # 15
n -= 3;  print(n)    # 12
n *= 2;  print(n)    # 24
n //= 5; print(n)    # 4
n **= 3; print(n)    # 64
n %= 10; print(n)    # 4

# Walrus operator: assign AND use in one expression
data = [1, 2, 3, 4, 5]
if (count := len(data)) > 3:
    print(f"List has {count} items")
```

### 5.5 Bitwise operators (work on the binary digits)

| Op | Name | Example (`a=12=1100`, `b=10=1010`) |
|---|---|---|
| `&` | AND | `a & b = 1000 = 8` |
| `\|` | OR | `a \| b = 1110 = 14` |
| `^` | XOR | `a ^ b = 0110 = 6` |
| `~` | NOT (invert) | `~a = -(a+1) = -13` |
| `<<` | left shift (×2ⁿ) | `a << 2 = 48` |
| `>>` | right shift (÷2ⁿ) | `a >> 2 = 3` |

```python
a, b = 12, 10
print(f"{a:04b} & {b:04b} = {a & b:04b} ({a & b})")
print(f"{a:04b} | {b:04b} = {a | b:04b} ({a | b})")
print(f"{a:04b} ^ {b:04b} = {a ^ b:04b} ({a ^ b})")
print(~a)
print(a << 2, a >> 2)

# Handy tricks
n = 37
print("odd" if n & 1 else "even")      # check parity with bitwise AND
print(n & (n - 1) == 0)                # is power of two?  False for 37
```

### 5.6 Identity operators: `is`, `is not`
Compare **identity** (same object in memory), not equality of value.

### 5.7 Membership operators: `in`, `not in`
Test whether a value is inside a sequence/collection.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)     # True  (equal values)
print(a is b)     # False (different objects)
print(a is c)     # True  (same object)

print(3 in a, 9 not in a)
print("ell" in "hello")          # substring test
print("k" in {"k": 1})           # for dicts it checks KEYS

# Small-integer caching (an interview classic)
p, q = 256, 256
print(p is q)              # True  (CPython caches -5..256)
r, s = 100000, 100000
print(r == s)              # always use == for values
```

## 6. Operator precedence (highest → lowest)

| Level | Operators |
|---|---|
| 1 | `()` parentheses |
| 2 | `**` |
| 3 | `+x`, `-x`, `~x` (unary) |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `<<`, `>>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `\|` |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `is not`, `in`, `not in` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |
| 14 | `=`, `+=`, ... (assignment) |

**Associativity:** most operators are left-to-right. `**` is **right-to-left**: `2 ** 3 ** 2 = 2 ** 9 = 512`.

**Memory trick:** *PEMDAS-ish* → **P**arentheses, **E**xponent, **M**ul/**D**iv, **A**dd/**S**ub, then comparisons, then `not > and > or`.

```python
print(2 + 3 * 4)          # 14
print((2 + 3) * 4)        # 20
print(2 ** 3 ** 2)        # 512  (right-associative)
print(-2 ** 2)            # -4   (** binds tighter than unary minus)
print((-2) ** 2)          # 4
print(10 - 4 - 3)         # 3    (left-to-right)
print(100 / 10 * 2)       # 20.0
print(not True or True)   # (not True) or True -> True
print(True or False and False)   # and before or -> True
print(5 > 3 > 1, 5 > 3 == 1)     # True False
```

```python
import math
print(math.pi, math.e)
print(math.sqrt(144), math.pow(2, 10), 2 ** 10)
print(math.factorial(6), math.gcd(48, 18), math.lcm(4, 6))
print(math.sin(math.radians(30)), math.cos(0))
print(math.log(100, 10), math.log2(8), math.log10(1000), math.log(math.e))
print(math.hypot(3, 4))
print(math.isqrt(50))            # integer square root
print(math.comb(5, 2), math.perm(5, 2))   # nCr, nPr
```

## 8. Mutable vs immutable — a first look

| Immutable (cannot change in place) | Mutable (can change in place) |
|---|---|
| `int`, `float`, `complex`, `bool`, `str`, `tuple`, `frozenset`, `bytes`, `None` | `list`, `dict`, `set`, `bytearray`, custom objects |

```python
s = "hello"
try:
    s[0] = "H"
except TypeError as e:
    print("TypeError:", e)

x = 5
print(id(x))
x += 1               # creates a NEW int object; x now points to it
print(id(x))

lst = [1, 2]
print(id(lst))
lst.append(3)        # same object modified in place
print(id(lst))
```

# 🕵️ Hidden Facts & Pro Tricks (Module 02)

```python
# 1. Chained assignment and chained comparison
a = b = c = 7
print(1 < a < 10 == 10 > 5)       # a whole chain of comparisons!

# 2. Underscore digit separators and any base
print(1_000_000 + 0b1111_0000)

# 3. Everything is an object — even numbers have methods
print((255).bit_length(), (10).to_bytes(2, "big"), (3.5).is_integer(), (4.0).is_integer())
print((1.25).as_integer_ratio())

# 4. Ints are cached: small ints (-5..256) are singletons (CPython detail)
x = 256; y = 256
print(x is y)                      # True
x = int("1000"); y = int("1000")
print(x is y, x == y)              # False True   <- why `is` is for None only!

# 5. bool is an int, so it can index and sum
print(sum([True, False, True, True]))         # count of Trues = 3
print(["no", "yes"][3 > 2])                   # poor man's ternary

# 6. NaN is not equal to itself (the only common value like that)
nan = float("nan")
print(nan == nan, nan != nan)

# 7. Floor division is floor, not truncate; use int() to truncate
print(-7 // 2, int(-7 / 2))

# 8. Rotate values without temp variables
a, b, c = 1, 2, 3
a, b, c = c, a, b
print(a, b, c)

# 9. Roots and pow with modulus
print(27 ** (1/3), pow(2, 100, 1000))         # pow(base, exp, mod) is a FAST modular exponent

# 10. Infinity is handy for min/max initialisation
lo = float("inf")
print(min(lo, 5))

# 11. Huge factorial with no overflow
import math
print(len(str(math.factorial(1000))), "digits in 1000!")

# 12. Complex numbers are built in
print((1 + 2j) ** 2, abs(3+4j))
```

**Hidden fact:** `0.1 + 0.2 != 0.3` is *not* a Python bug — every language using IEEE‑754 doubles (JavaScript, Java, C) does the same. The `fractions` module gives exact rational maths:

```python
from fractions import Fraction
print(Fraction(1, 10) + Fraction(2, 10) == Fraction(3, 10), Fraction(1, 3) + Fraction(1, 6))
```
