## 1. Taking input — `input()`

`input(prompt)` pauses the program, shows the prompt, waits for the user to type and press Enter, and **always returns a `str`**.

````python
name = input("Enter your name: ")
print("Hello,", name)
print(type(name))
````

### The #1 beginner bug: input is text!

````python
a = input("First number: ")    # '5'
b = input("Second number: ")   # '5'
print(a + b)                   # '55' – string concatenation, NOT addition
print(int(a) + int(b))         # 10   – convert first
````

### Reading numbers and multiple values

````python
age = int(input("Age: "))
price = float(input("Price: "))
print(age + 1, price * 2)

# Two values on ONE line, separated by spaces:
x, y = map(int, input("Enter two numbers: ").split())
print(x, y, x + y)

# A list of numbers on one line:
nums = list(map(int, input("Enter numbers: ").split()))
print(nums, sum(nums))
````

### Safe input with validation (try/except loop)
Real programs must survive bad input. Pattern:

````python
def read_int(prompt):
    while True:                       # keep asking until valid
        try:
            return int(input(prompt))
        except ValueError:
            print("  ❌ That's not a whole number. Try again.")

# (In this demo the fake input feeds '5', so it succeeds immediately.)
n = read_int("Enter a whole number: ")
print("You entered", n)
````

## 3. Conditional statements

Programs need to **make decisions**. Python evaluates a condition (an expression that is truthy or falsy) and runs a block if it is true.

### 3.1 `if`

```python
if condition:
    # runs only when condition is True
```

````python
temperature = 35
if temperature > 30:
    print("It's hot outside 🥵")
print("Program continues...")
````

### 3.2 `if – else`

````python
number = 7
if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
````

### 3.3 `if – elif – else` (ladder)
Conditions are tested **top to bottom**; the *first* true one runs and the rest are skipped. `else` catches everything else.

````python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks} -> Grade {grade}")
````

> ⚠️ **Order matters.** If you put `marks >= 40` first, an 85 would get "D" because it's checked first. Always order from most specific/strict to least.

### 3.4 Nested `if`

````python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Show your ID")
else:
    print("Too young")

# Same logic, flatter (preferred): combine with `and`
if age >= 18 and has_id:
    print("Entry allowed (flat version)")
````

### 3.5 Combining conditions: `and`, `or`, `not`, `in`, chained comparisons

````python
year = 2024
# Leap year: divisible by 4, but not by 100 unless also by 400
if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")

ch = "e"
if ch in "aeiou":
    print(ch, "is a vowel")

score = 55
if 40 <= score <= 100:
    print("Pass")

if not (score < 40):
    print("Not failing")
````

### 3.6 Ternary (conditional) expression
`value_if_true if condition else value_if_false`

````python
age = 17
status = "adult" if age >= 18 else "minor"
print(status)

n = -5
print("positive" if n > 0 else "negative" if n < 0 else "zero")   # nested ternary (use sparingly)
print(abs(n) if n < 0 else n)
````

### 3.7 Truthiness in conditions
You don't need `== True` or `len(x) > 0`. Use the value itself.

````python
items = []
if items:                    # empty list is falsy
    print("has items")
else:
    print("empty list")

name = ""
if not name:
    print("Name is missing")

# Pythonic vs non-Pythonic
flag = True
if flag == True: pass        # ❌ redundant
if flag: pass                # ✅

x = None
if x is None:                # ✅ use `is` for None
    print("x is None")
````

## 4. `match-case` (Python 3.10+)
Like `switch` in other languages but far more powerful (pattern matching).

````python
def http_status(code):
    match code:
        case 200:
            return "OK"
        case 301 | 302:            # OR pattern
            return "Redirect"
        case 404:
            return "Not Found"
        case c if c >= 500:        # guard
            return "Server error"
        case _:                    # wildcard (default)
            return "Unknown"

for c in (200, 302, 404, 503, 999):
    print(c, http_status(c))
````

````python
# Matching structures
def describe(point):
    match point:
        case (0, 0):
            return "origin"
        case (x, 0):
            return f"on x-axis at {x}"
        case (0, y):
            return f"on y-axis at {y}"
        case (x, y):
            return f"point ({x}, {y})"
        case _:
            return "not a point"

print(describe((0, 0)), "|", describe((3, 0)), "|", describe((2, 5)), "|", describe("hi"))
````

### Dictionary-dispatch: the classic pre-3.10 switch replacement

````python
def calc(op, a, b):
    ops = {
        "+": lambda: a + b,
        "-": lambda: a - b,
        "*": lambda: a * b,
        "/": lambda: a / b if b else "cannot divide by 0",
    }
    return ops.get(op, lambda: "invalid operator")()

print(calc("+", 6, 3), calc("/", 6, 0), calc("^", 1, 2))
````

## 5. Worked examples — study these patterns

````python
# 1. Largest of three numbers
a, b, c = 14, 27, 9
if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c
print("Largest:", largest, "| built-in:", max(a, b, c))

# 2. Triangle type from sides
def triangle_type(a, b, c):
    if a + b <= c or a + c <= b or b + c <= a:
        return "not a triangle"
    if a == b == c:
        return "equilateral"
    if a == b or b == c or a == c:
        return "isosceles"
    return "scalene"
print(triangle_type(3, 3, 3), triangle_type(3, 3, 5), triangle_type(3, 4, 5), triangle_type(1, 2, 8))

# 3. Character classifier
def classify(ch):
    if ch.isalpha():
        return "vowel" if ch.lower() in "aeiou" else "consonant"
    elif ch.isdigit():
        return "digit"
    elif ch.isspace():
        return "whitespace"
    return "special"
print([classify(c) for c in "a5 z$"])

# 4. BMI calculator
def bmi_category(weight_kg, height_m):
    bmi = weight_kg / height_m ** 2
    if bmi < 18.5:   cat = "Underweight"
    elif bmi < 25:   cat = "Normal"
    elif bmi < 30:   cat = "Overweight"
    else:            cat = "Obese"
    return round(bmi, 1), cat
print(bmi_category(70, 1.75))

# 5. Simple electricity bill (slab based)
def bill(units):
    if units <= 100:
        return units * 1.5
    elif units <= 200:
        return 100 * 1.5 + (units - 100) * 2.5
    else:
        return 100 * 1.5 + 100 * 2.5 + (units - 200) * 4
print(bill(80), bill(150), bill(250))

# 6. Quadratic roots
import math
def roots(a, b, c):
    d = b * b - 4 * a * c
    if d > 0:
        return ((-b + math.sqrt(d)) / (2 * a), (-b - math.sqrt(d)) / (2 * a))
    elif d == 0:
        return (-b / (2 * a),)
    return "complex roots"
print(roots(1, -3, 2), roots(1, 2, 1), roots(1, 0, 1))

# 7. Day of week via number
days = {1: "Mon", 2: "Tue", 3: "Wed", 4: "Thu", 5: "Fri", 6: "Sat", 7: "Sun"}
print(days.get(3, "invalid"), days.get(9, "invalid"))
````

## 6. Common mistakes

| Mistake | Fix |
|---|---|
| `if x = 5:` (assignment) | `if x == 5:` |
| Forgetting the colon `:` | `if x > 5:` |
| Wrong indentation | 4 spaces, consistently |
| `if x == 1 or 2:` (always True!) | `if x == 1 or x == 2:` or `if x in (1, 2):` |
| Comparing floats with `==` | `math.isclose()` |
| Using `input()` result as number | wrap with `int()` / `float()` |

````python
x = 5
print(x == 1 or 2)          # 2 -> truthy, so the condition would ALWAYS pass
print(x in (1, 2))          # False, the intended meaning
````

# 🕵️ Hidden Facts & Pro Tricks (Module 04)

````python
# 1. any() / all() replace long if-chains
marks = [45, 67, 89]
print("passed all" if all(m >= 40 for m in marks) else "some failed")
print("has distinction" if any(m >= 85 for m in marks) else "none")

# 2. Set membership beats chained ors
day = "Sat"
print("weekend" if day in {"Sat", "Sun"} else "weekday")

# 3. Booleans as indexes -> tiny lookup ternary
n = 7
print(["even", "odd"][n % 2])

# 4. Dict dispatch table: replace if/elif ladders
actions = {"start": lambda: "Starting...", "stop": lambda: "Stopping..."}
print(actions.get("start", lambda: "Unknown")())

# 5. Guard clauses flatten nested ifs
def can_vote(age, citizen):
    if age < 18:
        return False
    if not citizen:
        return False
    return True
print(can_vote(20, True), can_vote(17, True))

# 6. Range checking with chained comparison
temp = 22
print("comfortable" if 20 <= temp <= 26 else "adjust AC")

# 7. Default with `or` fails when 0 is a valid value
volume = 0
print(volume or 50)                            # 50 (BUG if 0 is valid!)
print(volume if volume is not None else 50)    # 0 (correct)

# 8. Walrus in conditions: compute once, use twice
import re
if m := re.search(r"\d+", "order 4521 shipped"):
    print("found", m.group())

# 9. Parse a line of numbers
a, b, c = map(int, "10 20 30".split())
print(a + b + c)

# 10. print with unpacking
print(*[1, 2, 3], sep=" + ", end=" = ")
print(sum([1, 2, 3]))

# Extra: getpass.getpass() hides typed passwords;
# sys.stdin.readline() is faster than input() for big competitive-programming inputs.
````

**Hidden fact — Python had no `switch` until 3.10** because Guido thought if/elif and dicts were enough. `match` is far more than a switch: it destructures lists, dicts and classes.

````python
def handle(cmd):
    match cmd.split():
        case ["go", direction]:
            return f"Going {direction}"
        case ["pick", "up", *items]:
            return f"Picked {', '.join(items)}"
        case ["quit"]:
            return "Bye"
        case _:
            return "Unknown command"
print(handle("go north"), "|", handle("pick up sword shield"), "|", handle("dance"))
````
