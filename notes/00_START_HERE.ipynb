## ⚙️ Setup (5 minutes)

1. Install Python 3.11+ from python.org (tick **"Add Python to PATH"**).
2. Install VS Code + the **Python** and **Jupyter** extensions (or run `pip install notebook` and `jupyter notebook`).
3. Optional libraries: `pip install numpy pandas matplotlib`.
4. Open a `.ipynb` file → choose the Python interpreter (top-right "Select Kernel") → press **Shift+Enter** to run cells.

Run the check below to make sure everything works:

```python
import sys, platform, importlib

print("Python  :", sys.version.split()[0], f"({platform.system()})")
assert sys.version_info >= (3, 10), "Please upgrade to Python 3.10 or newer (match-case needs 3.10+)."

for lib, why in [("numpy", "Module 18"), ("pandas", "Module 18, Project 5"), ("matplotlib", "charts (optional)")]:
    try:
        mod = importlib.import_module(lib)
        print(f"{lib:<11}: {mod.__version__}  ✅")
    except ImportError:
        print(f"{lib:<11}: not installed ❌  ->  pip install {lib}   ({why})")

print("\nHello, future Python developer! 🐍")
```

## 🧠 How to study so it actually sticks

1. **Type, don't paste.** Your fingers learn syntax that your eyes only recognise.
2. **Predict before you run.** Write the expected output as a comment, then compare. Every mismatch is a lesson.
3. **Break it.** Change a value, delete a colon, use the wrong type — read the error message. Errors are teachers.
4. **Explain it out loud** ("rubber duck"). If you can't explain a concept simply, you don't own it yet.
5. **Space it.** Revisit topics after 1 → 3 → 7 → 14 days (the `challenges/` notebooks are perfect for this).
6. **Build small things** every week. The projects folder is your portfolio.
7. **Keep a mistake log.** Bugs you've made once you won't make thrice.

## 🆘 When you're stuck
1. Read the **last line** of the error.
2. `help(thing)` / `dir(thing)` / `type(thing)` in a cell.
3. Search the exact error message (without your file paths).
4. Read the official tutorial: https://docs.python.org/3/tutorial/
5. Ask for help with a **minimal example** of what you tried and what you expected.
