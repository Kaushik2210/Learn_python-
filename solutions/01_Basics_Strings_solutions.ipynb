### 🟢 Problem 1 — Temperature converter
Write `c_to_f(c)` that converts Celsius to Fahrenheit: `F = C × 9/5 + 32`. Return a **float rounded to 1 decimal**.

<details><summary>💡 Hint</summary>Use `round(value, 1)`.</details>

```python
def c_to_f(c):
    return round(c * 9 / 5 + 32, 1)
```

```python
check("0C -> 32.0F", lambda: c_to_f(0) == 32.0)
check("100C -> 212.0F", lambda: c_to_f(100) == 212.0)
check("-40C -> -40.0F", lambda: c_to_f(-40) == -40.0)
check("36.6C -> 97.9F", lambda: c_to_f(36.6) == 97.9)
```

### 🟢 Problem 2 — Digit tools
Write `digit_sum(n)` returning the sum of the digits of a non-negative integer, using **arithmetic** (`%` and `//`) rather than converting to a string. Also write `is_even_digit_count(n)` returning `True` if `n` has an even number of digits.

<details><summary>💡 Hint</summary>`n % 10` is the last digit, `n // 10` removes it. For digit count you may use `len(str(n))`.</details>

```python
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total

def is_even_digit_count(n):
    return len(str(n)) % 2 == 0
```

```python
check("digit_sum(1234) == 10", lambda: digit_sum(1234) == 10)
check("digit_sum(0) == 0", lambda: digit_sum(0) == 0)
check("digit_sum(99999) == 45", lambda: digit_sum(99999) == 45)
check("even digit count 1234", lambda: is_even_digit_count(1234) is True)
check("odd digit count 123", lambda: is_even_digit_count(123) is False)
```

### 🟢 Problem 3 — Swap without a temp variable
Write `swap(a, b)` that returns the two values swapped as a tuple `(b, a)` using tuple unpacking (one line, no third variable).

```python
def swap(a, b):
    a, b = b, a
    return a, b
```

```python
check("swap ints", lambda: swap(1, 2) == (2, 1))
check("swap str/list", lambda: swap("x", [1]) == ([1], "x"))
```

### 🟢 Problem 4 — Reverse & palindrome
Write `is_palindrome(s)`: `True` if `s` reads the same forwards and backwards, **ignoring case, spaces and punctuation**.
`"A man, a plan, a canal: Panama"` → `True`.

<details><summary>💡 Hint</summary>Keep only `c.isalnum()` characters, lowercase them, compare with the reverse `[::-1]`.</details>

```python
def is_palindrome(s):
    cleaned = "".join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]
```

```python
check("panama", lambda: is_palindrome("A man, a plan, a canal: Panama") is True)
check("hello", lambda: is_palindrome("hello") is False)
check("empty string", lambda: is_palindrome("") is True)
check("single char", lambda: is_palindrome("x") is True)
check("Was it a car", lambda: is_palindrome("Was it a car or a cat I saw?") is True)
```

### 🟡 Problem 5 — Vowel/consonant/digit counter
Write `char_stats(s)` returning a dict `{"vowels": v, "consonants": c, "digits": d, "others": o}` (letters case-insensitive; spaces and punctuation count as "others").

```python
def char_stats(s):
    stats = {"vowels": 0, "consonants": 0, "digits": 0, "others": 0}
    for ch in s.lower():
        if ch in "aeiou":
            stats["vowels"] += 1
        elif ch.isalpha():
            stats["consonants"] += 1
        elif ch.isdigit():
            stats["digits"] += 1
        else:
            stats["others"] += 1
    return stats
```

```python
check("Hello World 123!", lambda: char_stats("Hello World 123!") == {"vowels": 3, "consonants": 7, "digits": 3, "others": 3})
check("empty", lambda: char_stats("") == {"vowels": 0, "consonants": 0, "digits": 0, "others": 0})
check("AEIOU", lambda: char_stats("AEIOU")["vowels"] == 5)
```

### 🟡 Problem 6 — Caesar cipher
Write `caesar(text, shift)` that shifts each **letter** by `shift` positions (wrapping around `z`→`a`), preserving case and leaving non-letters unchanged. A negative shift decrypts.

<details><summary>💡 Hint</summary>For lowercase: `chr((ord(c) - ord('a') + shift) % 26 + ord('a'))`. Do the same with `'A'` for uppercase.</details>

```python
def caesar(text, shift):
    out = []
    for c in text:
        if c.isalpha():
            base = ord("A") if c.isupper() else ord("a")
            out.append(chr((ord(c) - base + shift) % 26 + base))
        else:
            out.append(c)
    return "".join(out)
```

```python
check("shift 3", lambda: caesar("abc xyz", 3) == "def abc")
check("keeps case & punctuation", lambda: caesar("Hello, World!", 5) == "Mjqqt, Btwqi!")
check("decrypt with negative", lambda: caesar(caesar("Python 3.13", 7), -7) == "Python 3.13")
check("shift 26 is identity", lambda: caesar("Same", 26) == "Same")
```

### 🟡 Problem 7 — Anagram groups
Write `group_anagrams(words)` returning a list of groups (lists) of words that are anagrams of each other. The order of groups/words doesn't matter for the test (it sorts them).

`["listen", "silent", "enlist", "google", "gogole", "cat"]` → `[["listen","silent","enlist"], ["google","gogole"], ["cat"]]`

<details><summary>💡 Hint</summary>Use a dict whose key is `"".join(sorted(word))` (or a tuple) and whose value is a list.</details>

```python
def group_anagrams(words):
    groups = {}
    for w in words:
        groups.setdefault("".join(sorted(w)), []).append(w)
    return list(groups.values())
```

```python
def _norm(g): return sorted(sorted(x) for x in g)
check("basic groups", lambda: _norm(group_anagrams(["listen", "silent", "enlist", "google", "gogole", "cat"])) == _norm([["listen","silent","enlist"], ["google","gogole"], ["cat"]]))
check("empty list", lambda: group_anagrams([]) == [])
check("no anagrams", lambda: _norm(group_anagrams(["a", "b"])) == [["a"], ["b"]])
```

### 🟡 Problem 8 — Compress a string (run-length encoding)
`compress("aaabccdddd")` → `"a3b1c2d4"`. Write `compress(s)` and its inverse `decompress(s)` (counts can be multi-digit: `"x12"` → 12 x's).

<details><summary>💡 Hint</summary>Loop, count consecutive equal chars. For decompress use `re.findall(r"(\D)(\d+)", s)`.</details>

```python
import re

def compress(s):
    if not s:
        return ""
    out, count = [], 1
    for prev, cur in zip(s, s[1:]):
        if cur == prev:
            count += 1
        else:
            out.append(f"{prev}{count}")
            count = 1
    out.append(f"{s[-1]}{count}")
    return "".join(out)

def decompress(s):
    return "".join(ch * int(n) for ch, n in re.findall(r"(\D)(\d+)", s))
```

```python
check("compress basic", lambda: compress("aaabccdddd") == "a3b1c2d4")
check("compress empty", lambda: compress("") == "")
check("compress single", lambda: compress("z") == "z1")
check("decompress basic", lambda: decompress("a3b1c2d4") == "aaabccdddd")
check("round trip long run", lambda: decompress(compress("x" * 15 + "y")) == "x" * 15 + "y")
```

### 🟡 Problem 9 — Word statistics
Write `word_stats(sentence)` returning a tuple `(word_count, longest_word, average_word_length_rounded_to_2)`. Ignore punctuation `.,!?;:` attached to words. If several words tie for longest, return the **first**.

`"The quick brown fox jumps!"` → `(5, "quick", 4.2)`

```python
def word_stats(sentence):
    words = [w.strip(".,!?;:") for w in sentence.split()]
    words = [w for w in words if w]
    longest = max(words, key=len)          # max returns the first of equal maxima
    avg = round(sum(len(w) for w in words) / len(words), 2)
    return len(words), longest, avg
```

```python
check("quick brown fox", lambda: word_stats("The quick brown fox jumps!") == (5, "quick", 4.2))
check("punctuation stripped", lambda: word_stats("Hi, there.") == (2, "there", 3.5))
check("tie -> first", lambda: word_stats("aa bb cc")[1] == "aa")
```

### 🔴 Problem 10 — Validate a password
Write `password_strength(pw)` returning `"weak"`, `"medium"` or `"strong"`:
- **strong**: length ≥ 8 and contains a lowercase, an uppercase, a digit **and** a special character (non-alphanumeric)
- **medium**: length ≥ 6 and satisfies at least 3 of those 4 character classes
- otherwise **weak**

```python
def password_strength(pw):
    classes = [
        any(c.islower() for c in pw),
        any(c.isupper() for c in pw),
        any(c.isdigit() for c in pw),
        any(not c.isalnum() for c in pw),
    ]
    if len(pw) >= 8 and all(classes):
        return "strong"
    if len(pw) >= 6 and sum(classes) >= 3:
        return "medium"
    return "weak"
```

```python
check("strong", lambda: password_strength("P@ssw0rd!") == "strong")
check("medium (no symbol)", lambda: password_strength("Passw0rd") == "medium")
check("weak (short)", lambda: password_strength("aB1") == "weak")
check("weak (one class)", lambda: password_strength("abcdefghij") == "weak")
check("medium (short but 3 classes)", lambda: password_strength("aB1cde") == "medium")
```

### 🔴 Problem 11 — Roman numerals
Write `to_roman(n)` for `1 <= n <= 3999` and `from_roman(s)` as its inverse.
`1994` → `"MCMXCIV"`, `58` → `"LVIII"`.

<details><summary>💡 Hint</summary>Keep a list of `(value, symbol)` pairs from large to small including 900 (CM), 400 (CD), 90 (XC), 40 (XL), 9 (IX), 4 (IV). Greedily subtract. For `from_roman`, if a symbol is smaller than the next one, subtract it.</details>

```python
_PAIRS = [(1000, "M"), (900, "CM"), (500, "D"), (400, "CD"), (100, "C"), (90, "XC"),
          (50, "L"), (40, "XL"), (10, "X"), (9, "IX"), (5, "V"), (4, "IV"), (1, "I")]

def to_roman(n):
    out = []
    for value, symbol in _PAIRS:
        count, n = divmod(n, value)
        out.append(symbol * count)
    return "".join(out)

def from_roman(s):
    vals = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
    total = 0
    for cur, nxt in zip(s, s[1:] + " "):
        v = vals[cur]
        total += -v if nxt in vals and vals[nxt] > v else v
    return total
```

```python
check("to_roman 1994", lambda: to_roman(1994) == "MCMXCIV")
check("to_roman 58", lambda: to_roman(58) == "LVIII")
check("to_roman 3999", lambda: to_roman(3999) == "MMMCMXCIX")
check("from_roman 1994", lambda: from_roman("MCMXCIV") == 1994)
check("round trip 1..500", lambda: all(from_roman(to_roman(i)) == i for i in range(1, 501)))
```

### 🔴 Problem 12 — Number to words
Write `number_to_words(n)` for `0 <= n <= 999`. Examples: `0`→`"zero"`, `15`→`"fifteen"`, `42`→`"forty-two"`, `100`→`"one hundred"`, `999`→`"nine hundred ninety-nine"`, `305`→`"three hundred five"`.

```python
ONES = ["zero", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine", "ten",
        "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen", "seventeen", "eighteen", "nineteen"]
TENS = ["", "", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]

def number_to_words(n):
    if n < 20:
        return ONES[n]
    if n < 100:
        t, o = divmod(n, 10)
        return TENS[t] + (f"-{ONES[o]}" if o else "")
    h, rest = divmod(n, 100)
    return ONES[h] + " hundred" + (f" {number_to_words(rest)}" if rest else "")
```

```python
check("zero", lambda: number_to_words(0) == "zero")
check("fifteen", lambda: number_to_words(15) == "fifteen")
check("forty-two", lambda: number_to_words(42) == "forty-two")
check("one hundred", lambda: number_to_words(100) == "one hundred")
check("three hundred five", lambda: number_to_words(305) == "three hundred five")
check("nine hundred ninety-nine", lambda: number_to_words(999) == "nine hundred ninety-nine")
```
